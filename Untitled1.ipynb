```python

import pandas as pd
import numpy as np

# ==========================================
# 1. LOAD DATASET
# ==========================================

df = pd.read_csv("nwmp_model_ready.csv")

print("Original Dataset Shape:", df.shape)

# ==========================================
# 2. DEFINE IMPORTANT COLUMNS
# ==========================================

features = [
    "pH", "DO", "BOD", "Cond",
    "Nitrate", "COD", "TotalColi", "FecalColi",
    "Turbidity", "TotAlk", "Chloride", "Sulphate",
    "Hardness", "TDS", "TSS", "Fluoride"
]

target = "wqi_class"

required_columns = features + [target]

# Check required columns
missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ==========================================
# 3. REMOVE DUPLICATE RECORDS
# ==========================================

duplicate_count = df.duplicated().sum()

df = df.drop_duplicates().copy()

print("Duplicates Removed:", duplicate_count)

# ==========================================
# 4. CONVERT MEASUREMENTS TO NUMERIC
# ==========================================

for col in features:
    df[col] = pd.to_numeric(
        df[col], errors="coerce"
    )

# ==========================================
# 5. HANDLE INVALID TARGET VALUES
# ==========================================

valid_classes = [
    "Excellent",
    "Good",
    "Poor",
    "Very_poor",
    "Unfit"
]

df[target] = df[target].astype("string").str.strip()

invalid_target = (
    df[target].isna()
    | ~df[target].isin(valid_classes)
)

print("Invalid Target Records:", invalid_target.sum())

# Remove records without a valid classification
df = df.loc[~invalid_target].copy()

# ==========================================
# 6. CHECK MISSING VALUES
# ==========================================

print("\nMissing Values Before Imputation:")

print(df[features].isnull().sum())

# Do not calculate imputation values on the
# entire dataset. Missing values will be handled
# using training data only during model training.

# ==========================================
# 7. CHECK INFINITE VALUES
# ==========================================

df[features] = df[features].replace(
    [np.inf, -np.inf], np.nan
)

print("\nInfinite values converted to NaN.")

# ==========================================
# 8. IDENTIFY SUSPICIOUS RECORDS
# ==========================================

if "review_group" in df.columns:

    print("\nRecord Review Groups:")

    print(df["review_group"].value_counts())

    suspicious_records = df[
        df["review_group"] == "suspect_error"
    ]

    print(
        "\nSuspicious Records:",
        len(suspicious_records)
    )

# Preserve these records for investigation.
# Do not automatically delete extreme measurements.

# ==========================================
# 9. CHECK OUTLIERS USING IQR
# ==========================================

print("\nOutlier Summary:")

for col in features:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit)
        | (df[col] > upper_limit)
    ]

    print(f"{col}: {len(outliers)} potential outliers")

# Outliers are identified, NOT removed.

# ==========================================
# 10. CHECK CLASS DISTRIBUTION
# ==========================================

print("\nWater Quality Class Distribution:")

print(df[target].value_counts())

# ==========================================
# 11. CHECK MISSING DATA SUMMARY
# ==========================================

print("\nFinal Missing Values:")

print(df[features].isnull().sum())

# ==========================================
# 12. SAVE CLEANED DATASET
# ==========================================

df.to_csv("nwmp_cleaned.csv", index=False)

print("\n================================")
print("DATA CLEANING COMPLETED")
print("================================")

print("Final Dataset Shape:", df.shape)

print("Saved as: nwmp_cleaned.csv")
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
import warnings

warnings.filterwarnings("ignore")

# =====================================================
# 1. LOAD CLEANED DATASET
# =====================================================

df = pd.read_csv("nwmp_cleaned.csv")

OUTPUT_DIR = Path("EDA_Results")
OUTPUT_DIR.mkdir(exist_ok=True)

print("=" * 60)
print("WATER QUALITY DATASET - EXPLORATORY DATA ANALYSIS")
print("=" * 60)

print("\nDataset Shape:", df.shape)
print("\nColumn Names:\n", df.columns.tolist())

# =====================================================
# 2. DEFINE FEATURES AND TARGET
# =====================================================

features = [
    "pH", "DO", "BOD", "Cond",
    "Nitrate", "COD", "TotalColi", "FecalColi",
    "Turbidity", "TotAlk", "Chloride", "Sulphate",
    "Hardness", "TDS", "TSS", "Fluoride"
]

target = "wqi_class"

for col in features:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# =====================================================
# 3. BASIC DATASET INFORMATION
# =====================================================

print("\n1. DATASET INFORMATION")

df.info()

print("\nDataset Description:")
print(df[features].describe().round(3))

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# Save descriptive statistics
df[features].describe().T.to_csv(
    OUTPUT_DIR / "descriptive_statistics.csv"
)

# =====================================================
# 4. WATER QUALITY CLASS DISTRIBUTION
# =====================================================

print("\n2. WATER QUALITY CLASS DISTRIBUTION")

class_counts = df[target].value_counts()

print(class_counts)

plt.figure(figsize=(9, 5))

plt.bar(
    class_counts.index.astype(str),
    class_counts.values
)

plt.title("Water Quality Class Distribution")
plt.xlabel("Water Quality Class")
plt.ylabel("Number of Samples")
plt.xticks(rotation=20)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "01_class_distribution.png",
    dpi=300
)

plt.show()

# Class percentage
class_percentage = (
    df[target].value_counts(normalize=True) * 100
)

class_percentage.to_csv(
    OUTPUT_DIR / "class_percentages.csv"
)

# =====================================================
# 5. HISTOGRAMS OF ALL CHEMICAL PARAMETERS
# =====================================================

print("\n3. CHEMICAL PARAMETER DISTRIBUTIONS")

fig, axes = plt.subplots(
    4, 4,
    figsize=(18, 15)
)

axes = axes.flatten()

for i, col in enumerate(features):

    values = df[col].dropna()

    axes[i].hist(
        values,
        bins=30,
        edgecolor="black"
    )

    axes[i].set_title(col)
    axes[i].set_xlabel("Measurement")
    axes[i].set_ylabel("Frequency")

plt.suptitle(
    "Distribution of Water Quality Parameters",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "02_parameter_histograms.png",
    dpi=300
)

plt.show()

# =====================================================
# 6. BOXPLOTS FOR ALL PARAMETERS
# =====================================================

print("\n4. BOXPLOT ANALYSIS")

fig, axes = plt.subplots(
    4, 4,
    figsize=(18, 15)
)

axes = axes.flatten()

for i, col in enumerate(features):

    axes[i].boxplot(
        df[col].dropna(),
        vert=True
    )

    axes[i].set_title(col)
    axes[i].set_ylabel("Measurement")

plt.suptitle(
    "Boxplot Analysis of Chemical Parameters",
    fontsize=16
)

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "03_parameter_boxplots.png",
    dpi=300
)

plt.show()

# =====================================================
# 7. CORRELATION MATRIX
# =====================================================

print("\n5. CORRELATION ANALYSIS")

correlation = df[features].corr()

print(correlation.round(3))

correlation.to_csv(
    OUTPUT_DIR / "correlation_matrix.csv"
)

fig, ax = plt.subplots(figsize=(14, 11))

image = ax.imshow(
    correlation,
    cmap="coolwarm",
    vmin=-1,
    vmax=1
)

ax.set_xticks(range(len(features)))
ax.set_yticks(range(len(features)))

ax.set_xticklabels(
    features,
    rotation=90
)

ax.set_yticklabels(features)

for i in range(len(features)):
    for j in range(len(features)):

        ax.text(
            j, i,
            f"{correlation.iloc[i, j]:.2f}",
            ha="center",
            va="center",
            fontsize=7
        )

plt.colorbar(image, label="Correlation")

plt.title("Chemical Parameter Correlation Heatmap")

plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "04_correlation_heatmap.png",
    dpi=300
)

plt.show()

# =====================================================
# 8. STRONGEST CORRELATED PARAMETER PAIRS
# =====================================================

print("\n6. STRONGEST CORRELATIONS")

upper_triangle = correlation.where(
    np.triu(
        np.ones(correlation.shape),
        k=1
    ).astype(bool)
)

strong_correlations = (
    upper_triangle.stack()
    .sort_values(key=abs, ascending=False)
)

print(strong_correlations.head(15))

strong_correlations.to_csv(
    OUTPUT_DIR / "strong_correlations.csv"
)

# =====================================================
# 9. CLASS-WISE CHEMICAL ANALYSIS
# =====================================================

print("\n7. CLASS-WISE CHEMICAL MEASUREMENTS")

class_analysis = df.groupby(target)[features].mean()

print(class_analysis.round(3))

class_analysis.to_csv(
    OUTPUT_DIR / "class_wise_parameter_means.csv"
)

# Plot class-wise parameter averages
class_analysis.T.plot(
    kind="bar",
    figsize=(18, 7)
)

plt.title("Average Chemical Measurements by Water Quality Class")
plt.xlabel("Chemical Parameters")
plt.ylabel("Average Measurement")
plt.xticks(rotation=45)
plt.legend(title="Water Quality Class")
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "05_classwise_parameter_analysis.png",
    dpi=300
)

plt.show()

# =====================================================
# 10. YEAR-WISE WATER QUALITY ANALYSIS
# =====================================================

print("\n8. YEAR-WISE ANALYSIS")

if "year" in df.columns:

    df["year"] = pd.to_numeric(
        df["year"],
        errors="coerce"
    )

    year_counts = (
        df.groupby("year")
        .size()
        .sort_index()
    )

    print("\nRecords Per Year:")
    print(year_counts)

    year_counts.to_csv(
        OUTPUT_DIR / "yearwise_record_counts.csv"
    )

    plt.figure(figsize=(12, 5))

    plt.plot(
        year_counts.index,
        year_counts.values,
        marker="o"
    )

    plt.title("Water Quality Records by Year")
    plt.xlabel("Year")
    plt.ylabel("Number of Samples")
    plt.grid(True)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / "06_yearwise_records.png",
        dpi=300
    )

    plt.show()

    # Year-wise class distribution
    year_class = pd.crosstab(
        df["year"],
        df[target],
        normalize="index"
    ) * 100

    print("\nYear-wise Class Percentage:")
    print(year_class.round(2))

    year_class.to_csv(
        OUTPUT_DIR / "yearwise_class_distribution.csv"
    )

    year_class.plot(
        kind="line",
        marker="o",
        figsize=(12, 6)
    )

    plt.title("Water Quality Class Trends Over Years")
    plt.xlabel("Year")
    plt.ylabel("Percentage of Samples")
    plt.grid(True)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / "07_yearwise_class_trends.png",
        dpi=300
    )

    plt.show()

else:
    print("Year column not found.")

# =====================================================
# 11. STATION-WISE WATER QUALITY ANALYSIS
# =====================================================

print("\n9. STATION-WISE ANALYSIS")

station_col = None

for col in ["station_name", "station_id"]:

    if col in df.columns:
        station_col = col
        break

if station_col:

    station_counts = (
        df.groupby(station_col)
        .size()
        .sort_values(ascending=False)
    )

    print("\nTotal Monitoring Stations:",
          df[station_col].nunique())

    print("\nTop 15 Stations by Sample Count:")
    print(station_counts.head(15))

    station_counts.to_csv(
        OUTPUT_DIR / "station_record_counts.csv"
    )

    station_counts.head(15).plot(
        kind="bar",
        figsize=(12, 6)
    )

    plt.title("Top 15 Monitoring Stations")
    plt.xlabel("Station")
    plt.ylabel("Number of Samples")
    plt.xticks(rotation=45)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / "08_station_analysis.png",
        dpi=300
    )

    plt.show()

    # Station-wise water quality classes
    station_classes = pd.crosstab(
        df[station_col],
        df[target]
    )

    station_classes.to_csv(
        OUTPUT_DIR / "stationwise_quality_classes.csv"
    )

else:
    print("Station name or ID column not found.")

# =====================================================
# 12. OUTLIER DETECTION USING IQR
# =====================================================

print("\n10. OUTLIER ANALYSIS")

outlier_results = []

for col in features:

    values = df[col].dropna()

    Q1 = values.quantile(0.25)
    Q3 = values.quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = values[
        (values < lower) |
        (values > upper)
    ]

    outlier_results.append({
        "Parameter": col,
        "Q1": Q1,
        "Q3": Q3,
        "Lower Limit": lower,
        "Upper Limit": upper,
        "Outlier Count": len(outliers),
        "Outlier Percentage":
            len(outliers) / len(values) * 100
            if len(values) else 0
    })

outlier_df = pd.DataFrame(outlier_results)

print(outlier_df.round(3))

outlier_df.to_csv(
    OUTPUT_DIR / "outlier_summary.csv",
    index=False
)

plt.figure(figsize=(13, 6))

plt.bar(
    outlier_df["Parameter"],
    outlier_df["Outlier Count"]
)

plt.title("Potential Outliers in Chemical Parameters")
plt.xlabel("Parameter")
plt.ylabel("Outlier Count")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig(
    OUTPUT_DIR / "09_outlier_summary.png",
    dpi=300
)

plt.show()

# =====================================================
# 13. EXTREME AND SUSPECTED RECORD ANALYSIS
# =====================================================

print("\n11. EXTREME AND SUSPECTED RECORD ANALYSIS")

if "review_group" in df.columns:

    review_counts = df["review_group"].value_counts()

    print(review_counts)

    review_counts.to_csv(
        OUTPUT_DIR / "review_group_summary.csv"
    )

    review_counts.plot(
        kind="bar",
        figsize=(9, 5)
    )

    plt.title("Normal, Extreme and Suspected Records")
    plt.xlabel("Review Group")
    plt.ylabel("Number of Samples")
    plt.xticks(rotation=20)
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / "10_review_group_analysis.png",
        dpi=300
    )

    plt.show()

    # Save suspected records for manual inspection
    if "suspect_record" in df.columns:

        suspect = df[
            df["suspect_record"].astype(str).str.lower()
            .isin(["true", "1", "yes"])
        ]

    else:

        suspect = df[
            df["review_group"] == "suspect_error"
        ]

    suspect.to_csv(
        OUTPUT_DIR / "suspected_records_for_review.csv",
        index=False
    )

    print(
        "\nSuspected Records Saved:",
        len(suspect)
    )

    # Compare documented review groups and WQI classes
    review_class = pd.crosstab(
        df["review_group"],
        df[target]
    )

    review_class.to_csv(
        OUTPUT_DIR / "review_group_vs_class.csv"
    )

    print("\nReview Group vs Quality Class:")
    print(review_class)

# =====================================================
# 14. WQI VALUE ANALYSIS
# =====================================================

print("\n12. WATER QUALITY INDEX ANALYSIS")

if "wqi" in df.columns:

    df["wqi"] = pd.to_numeric(
        df["wqi"],
        errors="coerce"
    )

    print(df["wqi"].describe())

    plt.figure(figsize=(9, 5))

    plt.hist(
        df["wqi"].dropna(),
        bins=30,
        edgecolor="black"
    )

    plt.title("Water Quality Index Distribution")
    plt.xlabel("WQI")
    plt.ylabel("Frequency")
    plt.tight_layout()

    plt.savefig(
        OUTPUT_DIR / "11_wqi_distribution.png",
        dpi=300
    )

    plt.show()

    wqi_class_analysis = (
        df.groupby(target)["wqi"]
        .agg(["count", "mean", "median", "min", "max"])
    )

    print("\nWQI by Quality Class:")
    print(wqi_class_analysis)

    wqi_class_analysis.to_csv(
        OUTPUT_DIR / "wqi_class_statistics.csv"
    )

# =====================================================
# 15. SAVE FINAL EDA REPORT
# =====================================================

report = []

report.append("WATER QUALITY EDA SUMMARY")
report.append("=" * 50)

report.append(f"Total Records: {len(df)}")
report.append(f"Total Columns: {len(df.columns)}")
report.append(
    f"Unique Classes: {df[target].nunique()}"
)

report.append("\nClass Distribution:")
report.append(class_counts.to_string())

report.append("\nMissing Values:")
report.append(
    df[features].isnull().sum().to_string()
)

report.append("\nOutlier Summary:")
report.append(
    outlier_df.to_string(index=False)
)

if "review_group" in df.columns:

    report.append("\nReview Group Summary:")
    report.append(review_counts.to_string())

with open(
    OUTPUT_DIR / "EDA_Final_Report.txt",
    "w",
    encoding="utf-8"
) as file:

    file.write("\n".join(report))

# =====================================================
# 16. COMPLETION
# =====================================================

print("\n" + "=" * 60)
print("EXPLORATORY DATA ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)

print("\nAll reports and graphs saved in:")
print(OUTPUT_DIR.resolve())

print("\nGenerated files:")

for file in sorted(OUTPUT_DIR.iterdir()):
    print("-", file.name)
```

```python

# ============================================================
# WATER QUALITY CLASSIFICATION & ANOMALY DETECTION
# STEP 3: FEATURE SELECTION AND ML PREPROCESSING
# ============================================================

import pandas as pd
import numpy as np
import joblib
import json
import warnings

from pathlib import Path

from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler, LabelEncoder
from sklearn.model_selection import StratifiedGroupKFold, GroupShuffleSplit

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

INPUT_FILE = "nwmp_cleaned.csv"

OUTPUT_DIR = Path("ML_Preprocessing")
OUTPUT_DIR.mkdir(exist_ok=True)

TEST_SIZE = 0.20
RANDOM_STATE = 42

print("=" * 65)
print("WATER QUALITY ML PREPROCESSING")
print("=" * 65)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(INPUT_FILE)

print("\nOriginal Dataset Shape:", df.shape)

# ============================================================
# 3. SELECT ORIGINAL CHEMICAL FEATURES
# ============================================================

features = [
    "pH",
    "DO",
    "BOD",
    "Cond",
    "Nitrate",
    "COD",
    "TotalColi",
    "FecalColi",
    "Turbidity",
    "TotAlk",
    "Chloride",
    "Sulphate",
    "Hardness",
    "TDS",
    "TSS",
    "Fluoride"
]

target = "wqi_class"
group_column = "station_id"

required_columns = features + [
    target,
    group_column
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing required columns: {missing_columns}"
    )

# ============================================================
# 4. DATA VALIDATION
# ============================================================

# Convert features into numeric format

for col in features:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

# Convert infinite values to missing values

df[features] = df[features].replace(
    [np.inf, -np.inf],
    np.nan
)

# Remove records without a target or station ID

before = len(df)

df = df.dropna(
    subset=[target, group_column]
).copy()

df[target] = df[target].astype(str).str.strip()

df[group_column] = df[group_column].astype(str)

print("\nRecords removed due to missing target/group:",
      before - len(df))

print("Remaining Records:", len(df))

# ============================================================
# 5. FEATURE AND TARGET SEPARATION
# ============================================================

X = df[features].copy()
y = df[target].copy()

groups = df[group_column].copy()

print("\nSelected Features:", len(features))

print("\nFeature Columns:")
print(features)

print("\nTarget Column:", target)

print("\nTarget Distribution:")
print(y.value_counts())

# ============================================================
# 6. ENCODE TARGET LABELS
# ============================================================

label_encoder = LabelEncoder()

y_encoded = pd.Series(
    label_encoder.fit_transform(y),
    index=y.index,
    name="target_encoded"
)

class_mapping = {
    str(label): int(encoded)
    for label, encoded in zip(
        label_encoder.classes_,
        range(len(label_encoder.classes_))
    )
}

print("\nClass Mapping:")
print(class_mapping)

# Save class mapping

with open(
    OUTPUT_DIR / "class_mapping.json",
    "w"
) as f:

    json.dump(
        class_mapping,
        f,
        indent=4
    )

joblib.dump(
    label_encoder,
    OUTPUT_DIR / "label_encoder.pkl"
)

# ============================================================
# 7. GROUP-AWARE TRAIN TEST SPLIT
# ============================================================

print("\nPerforming Station-Aware Train-Test Split...")

unique_groups = groups.nunique()

if unique_groups < 2:
    raise ValueError(
        "At least two distinct stations are required."
    )

# Try stratified group splitting.
# Select the fold with the closest class distribution
# and approximately 20% test records.

n_splits = min(5, unique_groups)

splitter = StratifiedGroupKFold(
    n_splits=n_splits,
    shuffle=True,
    random_state=RANDOM_STATE
)

overall_distribution = (
    y.value_counts(normalize=True)
)

best_score = float("inf")
best_split = None

for train_idx, test_idx in splitter.split(
    X, y, groups
):

    y_train_candidate = y.iloc[train_idx]
    y_test_candidate = y.iloc[test_idx]

    test_fraction = len(test_idx) / len(df)

    test_distribution = (
        y_test_candidate.value_counts(normalize=True)
    )

    distribution_error = sum(
        abs(
            overall_distribution.get(cls, 0)
            - test_distribution.get(cls, 0)
        )
        for cls in overall_distribution.index
    )

    missing_classes = len(
        set(y.unique()) - set(y_test_candidate.unique())
    )

    missing_train_classes = len(
        set(y.unique()) - set(y_train_candidate.unique())
    )

    score = (
        abs(test_fraction - TEST_SIZE)
        + distribution_error
        + 2.0 * missing_classes
        + 5.0 * missing_train_classes
    )

    if score < best_score:
        best_score = score
        best_split = (train_idx, test_idx)

train_idx, test_idx = best_split

X_train = X.iloc[train_idx].copy()
X_test = X.iloc[test_idx].copy()

y_train = y_encoded.iloc[train_idx].copy()
y_test = y_encoded.iloc[test_idx].copy()

train_groups = groups.iloc[train_idx]
test_groups = groups.iloc[test_idx]

print("\nTrain Records:", len(X_train))
print("Test Records:", len(X_test))

print("Train Stations:", train_groups.nunique())
print("Test Stations:", test_groups.nunique())

# Verify no station overlap

overlap = set(train_groups) & set(test_groups)

print("Overlapping Stations:", len(overlap))

if overlap:
    raise RuntimeError(
        "Station leakage detected!"
    )

# ============================================================
# 8. CHECK CLASS DISTRIBUTION AFTER SPLITTING
# ============================================================

print("\nTraining Class Distribution:")

print(
    label_encoder.inverse_transform(
        y_train.to_numpy()
    )
)

train_class_counts = pd.Series(
    label_encoder.inverse_transform(
        y_train.to_numpy()
    )
).value_counts()

test_class_counts = pd.Series(
    label_encoder.inverse_transform(
        y_test.to_numpy()
    )
).value_counts()

print("\nTraining Class Counts:")
print(train_class_counts)

print("\nTesting Class Counts:")
print(test_class_counts)

missing_test_classes = (
    set(label_encoder.classes_)
    - set(test_class_counts.index)
)

missing_train_classes = (
    set(label_encoder.classes_)
    - set(train_class_counts.index)
)

if missing_test_classes:
    print(
        "\nWARNING: Classes absent from test set:",
        missing_test_classes
    )

if missing_train_classes:
    raise ValueError(
        "Some classes are missing from training data: "
        f"{missing_train_classes}. "
        "Revise the split before training."
    )

# ============================================================
# 9. SUPERVISED LEARNING PREPROCESSING
# ============================================================

print("\nPreparing Random Forest Input...")

# Random Forest does not require feature scaling.
# Median imputation is fitted on training data only.

rf_imputer = SimpleImputer(
    strategy="median"
)

X_train_rf = pd.DataFrame(
    rf_imputer.fit_transform(X_train),
    columns=features,
    index=X_train.index
)

X_test_rf = pd.DataFrame(
    rf_imputer.transform(X_test),
    columns=features,
    index=X_test.index
)

joblib.dump(
    rf_imputer,
    OUTPUT_DIR / "rf_imputer.pkl"
)

# ============================================================
# 10. UNSUPERVISED LEARNING PREPROCESSING
# ============================================================

print("\nPreparing Isolation Forest Input...")

# Separate imputer and scaler.
# Both are fitted only on training measurements.

if_imputer = SimpleImputer(
    strategy="median"
)

X_train_if_imputed = pd.DataFrame(
    if_imputer.fit_transform(X_train),
    columns=features,
    index=X_train.index
)

X_test_if_imputed = pd.DataFrame(
    if_imputer.transform(X_test),
    columns=features,
    index=X_test.index
)

if_scaler = RobustScaler()

X_train_if = pd.DataFrame(
    if_scaler.fit_transform(X_train_if_imputed),
    columns=features,
    index=X_train.index
)

X_test_if = pd.DataFrame(
    if_scaler.transform(X_test_if_imputed),
    columns=features,
    index=X_test.index
)

joblib.dump(
    if_imputer,
    OUTPUT_DIR / "if_imputer.pkl"
)

joblib.dump(
    if_scaler,
    OUTPUT_DIR / "if_scaler.pkl"
)

# ============================================================
# 11. SAVE TRAINING AND TESTING DATASETS
# ============================================================

# Random Forest input

X_train_rf.to_csv(
    OUTPUT_DIR / "X_train_rf.csv",
    index=False
)

X_test_rf.to_csv(
    OUTPUT_DIR / "X_test_rf.csv",
    index=False
)

# Isolation Forest input

X_train_if.to_csv(
    OUTPUT_DIR / "X_train_if.csv",
    index=False
)

X_test_if.to_csv(
    OUTPUT_DIR / "X_test_if.csv",
    index=False
)

# Target values

y_train.to_csv(
    OUTPUT_DIR / "y_train.csv",
    index=False
)

y_test.to_csv(
    OUTPUT_DIR / "y_test.csv",
    index=False
)

# ============================================================
# 12. SAVE TEST METADATA
# ============================================================

metadata_columns = [
    col for col in [
        "station_id",
        "station_name",
        "source_type",
        "year",
        "wqi_class",
        "review_group",
        "extreme_flag",
        "suspect_record",
        "suggested_split"
    ]
    if col in df.columns
]

test_metadata = df.loc[
    X_test.index,
    metadata_columns
].copy()

test_metadata.to_csv(
    OUTPUT_DIR / "test_metadata.csv",
    index=False
)

# Save training metadata too

train_metadata = df.loc[
    X_train.index,
    metadata_columns
].copy()

train_metadata.to_csv(
    OUTPUT_DIR / "train_metadata.csv",
    index=False
)

# ============================================================
# 13. SAVE FEATURE CONFIGURATION
# ============================================================

configuration = {
    "features": features,
    "target": target,
    "group_column": group_column,
    "number_of_features": len(features),
    "number_of_records": len(df),
    "train_records": len(X_train),
    "test_records": len(X_test),
    "train_stations": int(train_groups.nunique()),
    "test_stations": int(test_groups.nunique()),
    "test_size": TEST_SIZE,
    "random_state": RANDOM_STATE,
    "classes": list(label_encoder.classes_),
    "class_mapping": class_mapping
}

with open(
    OUTPUT_DIR / "model_configuration.json",
    "w"
) as f:

    json.dump(
        configuration,
        f,
        indent=4
    )

# ============================================================
# 14. PREPROCESSING SUMMARY
# ============================================================

summary = []

summary.append("WATER QUALITY ML PREPROCESSING REPORT")
summary.append("=" * 55)

summary.append(
    f"Total records: {len(df)}"
)

summary.append(
    f"Total features: {len(features)}"
)

summary.append(
    f"Training records: {len(X_train)}"
)

summary.append(
    f"Testing records: {len(X_test)}"
)

summary.append(
    f"Training stations: {train_groups.nunique()}"
)

summary.append(
    f"Testing stations: {test_groups.nunique()}"
)

summary.append(
    f"Station overlap: {len(overlap)}"
)

summary.append(
    f"Target classes: {list(label_encoder.classes_)}"
)

summary.append(
    f"Missing training measurements: {X_train.isnull().sum().sum()}"
)

summary.append(
    f"Missing test measurements: {X_test.isnull().sum().sum()}"
)

summary.append("\nTraining Class Counts:")
summary.append(train_class_counts.to_string())

summary.append("\nTesting Class Counts:")
summary.append(test_class_counts.to_string())

summary.append("\nSelected Features:")
summary.append(", ".join(features))

with open(
    OUTPUT_DIR / "preprocessing_report.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write("\n".join(summary))

# ============================================================
# 15. FINAL VALIDATION
# ============================================================

assert len(X_train) + len(X_test) == len(df)

assert X_train_rf.shape[1] == len(features)

assert X_train_if.shape[1] == len(features)

assert len(overlap) == 0

assert not np.isnan(X_train_rf.to_numpy()).any()

assert not np.isnan(X_test_rf.to_numpy()).any()

assert not np.isnan(X_train_if.to_numpy()).any()

assert not np.isnan(X_test_if.to_numpy()).any()

print("\n" + "=" * 65)
print("PREPROCESSING COMPLETED SUCCESSFULLY!")
print("=" * 65)

print("\nRandom Forest Training Shape:",
      X_train_rf.shape)

print("Random Forest Testing Shape:",
      X_test_rf.shape)

print("\nIsolation Forest Training Shape:",
      X_train_if.shape)

print("Isolation Forest Testing Shape:",
      X_test_if.shape)

print("\nStation Overlap:", len(overlap))

print("\nAll files saved in:")
print(OUTPUT_DIR.resolve())

print("\nNEXT: RANDOM FOREST CLASSIFICATION")
```

```python

# ============================================================
# WATER QUALITY PROJECT
# STEP 4: SUPERVISED MACHINE LEARNING
# RANDOM FOREST CLASSIFICATION
# ============================================================

import pandas as pd
import numpy as np
import joblib
import json
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score,
    ConfusionMatrixDisplay
)
from sklearn.inspection import permutation_importance

# ============================================================
# 1. CONFIGURATION
# ============================================================

INPUT_DIR = Path("ML_Preprocessing")
OUTPUT_DIR = Path("Supervised_ML")

OUTPUT_DIR.mkdir(exist_ok=True)

RANDOM_STATE = 42

print("=" * 65)
print("WATER QUALITY - RANDOM FOREST CLASSIFICATION")
print("=" * 65)

# ============================================================
# 2. LOAD PREPROCESSED DATA
# ============================================================

X_train = pd.read_csv(
    INPUT_DIR / "X_train_rf.csv"
)

X_test = pd.read_csv(
    INPUT_DIR / "X_test_rf.csv"
)

y_train = pd.read_csv(
    INPUT_DIR / "y_train.csv"
).iloc[:, 0]

y_test = pd.read_csv(
    INPUT_DIR / "y_test.csv"
).iloc[:, 0]

label_encoder = joblib.load(
    INPUT_DIR / "label_encoder.pkl"
)

features = X_train.columns.tolist()

class_names = label_encoder.classes_

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

print("\nClasses:", list(class_names))

# ============================================================
# 3. INITIALIZE RANDOM FOREST
# ============================================================

rf_model = RandomForestClassifier(

    n_estimators=300,

    max_depth=None,

    min_samples_split=4,

    min_samples_leaf=2,

    class_weight="balanced_subsample",

    random_state=RANDOM_STATE,

    n_jobs=-1

)

# ============================================================
# 4. TRAIN MODEL
# ============================================================

print("\nTraining Random Forest Model...")

rf_model.fit(
    X_train,
    y_train
)

print("Training Completed Successfully!")

# ============================================================
# 5. PREDICTION
# ============================================================

y_pred = rf_model.predict(X_test)

# Decode predictions

actual_labels = label_encoder.inverse_transform(
    y_test.to_numpy().astype(int)
)

predicted_labels = label_encoder.inverse_transform(
    y_pred.astype(int)
)

# ============================================================
# 6. MODEL EVALUATION
# ============================================================

accuracy = accuracy_score(
    y_test,
    y_pred
)

balanced_accuracy = balanced_accuracy_score(
    y_test,
    y_pred
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\n" + "=" * 65)
print("MODEL PERFORMANCE")
print("=" * 65)

print(f"Accuracy: {accuracy:.4f}")

print(f"Balanced Accuracy: {balanced_accuracy:.4f}")

print(f"Macro F1 Score: {macro_f1:.4f}")

print(f"Weighted F1 Score: {weighted_f1:.4f}")

# ============================================================
# 7. CLASSIFICATION REPORT
# ============================================================

report = classification_report(

    y_test,

    y_pred,

    labels=np.arange(len(class_names)),

    target_names=class_names,

    output_dict=True,

    zero_division=0

)

report_df = pd.DataFrame(report).transpose()

print("\nCLASSIFICATION REPORT")

print(report_df.round(4))

report_df.to_csv(
    OUTPUT_DIR / "classification_report.csv"
)

# ============================================================
# 8. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    y_pred,

    labels=np.arange(len(class_names))

)

cm_df = pd.DataFrame(

    cm,

    index=class_names,

    columns=class_names

)

print("\nCONFUSION MATRIX")

print(cm_df)

cm_df.to_csv(
    OUTPUT_DIR / "confusion_matrix.csv"
)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=class_names

)

fig, ax = plt.subplots(figsize=(9, 7))

disp.plot(

    cmap="Blues",

    values_format="d",

    ax=ax,

    xticks_rotation=45

)

plt.title("Random Forest - Water Quality Classification")

plt.tight_layout()

plt.savefig(

    OUTPUT_DIR / "confusion_matrix.png",

    dpi=300

)

plt.show()

# ============================================================
# 9. FEATURE IMPORTANCE
# ============================================================

importance_df = pd.DataFrame({

    "Feature": features,

    "Importance": rf_model.feature_importances_

}).sort_values(

    by="Importance",

    ascending=False

)

print("\nFEATURE IMPORTANCE")

print(importance_df)

importance_df.to_csv(

    OUTPUT_DIR / "feature_importance.csv",

    index=False

)

plt.figure(figsize=(11, 7))

plt.barh(

    importance_df["Feature"],

    importance_df["Importance"]

)

plt.gca().invert_yaxis()

plt.title("Random Forest - Feature Importance")

plt.xlabel("Importance")

plt.tight_layout()

plt.savefig(

    OUTPUT_DIR / "feature_importance.png",

    dpi=300

)

plt.show()

# ============================================================
# 10. PERMUTATION IMPORTANCE
# ============================================================

print("\nCalculating Permutation Importance...")

permutation = permutation_importance(

    rf_model,

    X_test,

    y_test,

    n_repeats=10,

    random_state=RANDOM_STATE,

    scoring="f1_macro",

    n_jobs=-1

)

permutation_df = pd.DataFrame({

    "Feature": features,

    "Importance_Mean": permutation.importances_mean,

    "Importance_Std": permutation.importances_std

}).sort_values(

    by="Importance_Mean",

    ascending=False

)

print("\nPERMUTATION IMPORTANCE")

print(permutation_df)

permutation_df.to_csv(

    OUTPUT_DIR / "permutation_importance.csv",

    index=False

)

# ============================================================
# 11. SAVE PREDICTION RESULTS
# ============================================================

prediction_results = pd.DataFrame({

    "Actual_Quality": actual_labels,

    "Predicted_Quality": predicted_labels,

    "Correct_Prediction": (
        actual_labels == predicted_labels
    )

})

prediction_results.to_csv(

    OUTPUT_DIR / "prediction_results.csv",

    index=False

)

# ============================================================
# 12. SAVE TRAINED MODEL
# ============================================================

joblib.dump(

    rf_model,

    OUTPUT_DIR / "random_forest_model.pkl"

)

# ============================================================
# 13. SAVE MODEL CONFIGURATION
# ============================================================

configuration = {

    "model_name": "RandomForestClassifier",

    "features": features,

    "classes": list(class_names),

    "n_estimators": 300,

    "random_state": RANDOM_STATE,

    "accuracy": float(accuracy),

    "balanced_accuracy": float(balanced_accuracy),

    "macro_f1": float(macro_f1),

    "weighted_f1": float(weighted_f1)

}

with open(

    OUTPUT_DIR / "model_configuration.json",

    "w"

) as file:

    json.dump(

        configuration,

        file,

        indent=4

    )

# ============================================================
# 14. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 65)

print("RANDOM FOREST MODEL COMPLETED!")

print("=" * 65)

print("\nAccuracy:", round(accuracy * 100, 2), "%")

print("Balanced Accuracy:", round(balanced_accuracy * 100, 2), "%")

print("Macro F1:", round(macro_f1, 4))

print("\nModel Saved At:")

print(
    (OUTPUT_DIR / "random_forest_model.pkl").resolve()
)

print("\nAll results saved in:")

print(OUTPUT_DIR.resolve())
 
```

```python

# ============================================================
# WATER QUALITY PROJECT
# STEP 4B: MULTIPLE CLASSIFICATION MODEL COMPARISON
# ============================================================

import pandas as pd
import numpy as np
import joblib
import time
import json
import matplotlib.pyplot as plt

from pathlib import Path

from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import (
    RandomForestClassifier,
    ExtraTreesClassifier,
    GradientBoostingClassifier
)
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import GroupKFold, cross_validate
from sklearn.metrics import (
    make_scorer,
    f1_score,
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    ConfusionMatrixDisplay,
    confusion_matrix
)

# ============================================================
# 1. CONFIGURATION
# ============================================================

INPUT_DIR = Path("ML_Preprocessing")
OUTPUT_DIR = Path("Model_Comparison")

OUTPUT_DIR.mkdir(exist_ok=True)

RANDOM_STATE = 42

print("=" * 70)
print("WATER QUALITY - MULTIPLE CLASSIFICATION MODEL COMPARISON")
print("=" * 70)

# ============================================================
# 2. LOAD PREPROCESSED DATA
# ============================================================

X_train = pd.read_csv(
    INPUT_DIR / "X_train_rf.csv"
)

X_test = pd.read_csv(
    INPUT_DIR / "X_test_rf.csv"
)

y_train = pd.read_csv(
    INPUT_DIR / "y_train.csv"
).iloc[:, 0]

y_test = pd.read_csv(
    INPUT_DIR / "y_test.csv"
).iloc[:, 0]

train_metadata = pd.read_csv(
    INPUT_DIR / "train_metadata.csv"
)

label_encoder = joblib.load(
    INPUT_DIR / "label_encoder.pkl"
)

class_names = label_encoder.classes_

all_labels = np.arange(len(class_names))

# Validate input alignment

assert len(X_train) == len(y_train)
assert len(X_test) == len(y_test)
assert len(train_metadata) == len(X_train)

assert list(X_train.columns) == list(X_test.columns)

features = X_train.columns.tolist()

print("\nTraining Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
print("Classes:", list(class_names))

# ============================================================
# 3. GROUP-AWARE CROSS VALIDATION
# ============================================================

if "station_id" not in train_metadata.columns:
    raise ValueError("station_id missing from training metadata")

groups = train_metadata["station_id"]

n_splits = min(5, groups.nunique())

if n_splits < 2:
    raise ValueError("Not enough unique stations for cross-validation")

cv = GroupKFold(n_splits=n_splits)

print("\nCross Validation:", n_splits, "Folds")
print("Unique Training Stations:", groups.nunique())

# Fixed-label macro F1 includes all five target classes.

macro_f1_scorer = make_scorer(
    f1_score,
    labels=all_labels,
    average="macro",
    zero_division=0
)

scoring = {
    "macro_f1": macro_f1_scorer,
    "balanced_accuracy": "balanced_accuracy",
    "accuracy": "accuracy"
}

# ============================================================
# 4. DEFINE CLASSIFICATION MODELS
# ============================================================

models = {

    "Dummy Baseline": DummyClassifier(
        strategy="most_frequent"
    ),

    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=3000,
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "Decision Tree": DecisionTreeClassifier(
        max_depth=8,
        min_samples_split=4,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=300,
        max_depth=None,
        min_samples_split=4,
        min_samples_leaf=2,
        class_weight="balanced_subsample",
        random_state=RANDOM_STATE,
        n_jobs=1
    ),

    "Extra Trees": ExtraTreesClassifier(
        n_estimators=300,
        min_samples_split=4,
        min_samples_leaf=2,
        class_weight="balanced",
        random_state=RANDOM_STATE,
        n_jobs=1
    ),

    "Support Vector Machine": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", SVC(
            kernel="rbf",
            C=1.0,
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "K-Nearest Neighbors": Pipeline([
        ("scaler", StandardScaler()),
        ("classifier", KNeighborsClassifier(
            n_neighbors=5,
            weights="distance"
        ))
    ]),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=150,
        learning_rate=0.05,
        max_depth=2,
        min_samples_leaf=2,
        random_state=RANDOM_STATE
    )
}

# ============================================================
# 5. CROSS VALIDATION MODEL EVALUATION
# ============================================================

cv_results = []

trained_models = {}

print("\nSTARTING MODEL CROSS VALIDATION")

for model_name, model in models.items():

    print("\n" + "-" * 60)
    print("Evaluating:", model_name)

    start_time = time.time()

    scores = cross_validate(

        model,

        X_train,

        y_train,

        groups=groups,

        cv=cv,

        scoring=scoring,

        n_jobs=-1,

        return_train_score=False,

        error_score="raise"

    )

    elapsed_time = time.time() - start_time

    cv_results.append({

        "Model": model_name,

        "CV_Macro_F1_Mean":
            scores["test_macro_f1"].mean(),

        "CV_Macro_F1_Std":
            scores["test_macro_f1"].std(),

        "CV_Balanced_Accuracy":
            scores["test_balanced_accuracy"].mean(),

        "CV_Accuracy":
            scores["test_accuracy"].mean(),

        "CV_Time_Seconds":
            elapsed_time

    })

    print(
        "Macro F1:",
        round(scores["test_macro_f1"].mean(), 4)
    )

    print(
        "Balanced Accuracy:",
        round(scores["test_balanced_accuracy"].mean(), 4)
    )

    print(
        "CV Time:",
        round(elapsed_time, 2),
        "seconds"
    )

    # Save fitted model on the complete training data

    model.fit(X_train, y_train)

    trained_models[model_name] = model

    safe_name = (
        model_name.lower()
        .replace(" ", "_")
        .replace("-", "_")
    )

    joblib.dump(

        model,

        OUTPUT_DIR / f"{safe_name}_model.pkl"

    )

# ============================================================
# 6. CROSS VALIDATION COMPARISON TABLE
# ============================================================

cv_df = pd.DataFrame(cv_results)

cv_df = cv_df.sort_values(

    by="CV_Macro_F1_Mean",

    ascending=False

)

cv_df.to_csv(

    OUTPUT_DIR / "cross_validation_comparison.csv",

    index=False

)

print("\n" + "=" * 70)

print("CROSS VALIDATION RESULTS")

print("=" * 70)

print(cv_df.round(4).to_string(index=False))

# ============================================================
# 7. SELECT BEST MODEL USING CROSS VALIDATION
# ============================================================

best_model_name = cv_df.iloc[0]["Model"]

best_model = trained_models[best_model_name]

print("\nBEST MODEL ACCORDING TO CV MACRO F1:")

print(best_model_name)

# ============================================================
# 8. FINAL HOLDOUT TEST COMPARISON
# ============================================================

# This evaluates all models on the same held-out test set.
# Use CV as the primary selection criterion, not this table,
# to avoid repeatedly selecting models against the test set.

holdout_results = []

for model_name, model in trained_models.items():

    predictions = model.predict(X_test)

    holdout_results.append({

        "Model": model_name,

        "Accuracy": accuracy_score(
            y_test, predictions
        ),

        "Balanced_Accuracy": balanced_accuracy_score(
            y_test, predictions
        ),

        "Macro_F1": f1_score(
            y_test,
            predictions,
            labels=all_labels,
            average="macro",
            zero_division=0
        ),

        "Weighted_F1": f1_score(
            y_test,
            predictions,
            average="weighted",
            zero_division=0
        )

    })

holdout_df = pd.DataFrame(holdout_results)

holdout_df = holdout_df.sort_values(

    by="Macro_F1",

    ascending=False

)

holdout_df.to_csv(

    OUTPUT_DIR / "holdout_test_comparison.csv",

    index=False

)

print("\n" + "=" * 70)

print("FINAL HOLDOUT TEST COMPARISON")

print("=" * 70)

print(holdout_df.round(4).to_string(index=False))

# ============================================================
# 9. VISUALIZE CROSS VALIDATION MACRO F1
# ============================================================

plot_df = cv_df.sort_values(
    "CV_Macro_F1_Mean",
    ascending=True
)

plt.figure(figsize=(11, 7))

plt.barh(

    plot_df["Model"],

    plot_df["CV_Macro_F1_Mean"],

    xerr=plot_df["CV_Macro_F1_Std"],

    capsize=4

)

plt.xlabel("Cross-Validation Macro F1")

plt.title("Classification Model Comparison")

plt.tight_layout()

plt.savefig(

    OUTPUT_DIR / "model_comparison_macro_f1.png",

    dpi=300

)

plt.show()

# ============================================================
# 10. FINAL BEST MODEL CLASSIFICATION REPORT
# ============================================================

best_predictions = best_model.predict(X_test)

best_report = classification_report(

    y_test,

    best_predictions,

    labels=all_labels,

    target_names=class_names,

    output_dict=True,

    zero_division=0

)

best_report_df = pd.DataFrame(best_report).transpose()

best_report_df.to_csv(

    OUTPUT_DIR / "best_model_classification_report.csv"

)

print("\nBEST MODEL CLASSIFICATION REPORT")

print(best_report_df.round(4))

# ============================================================
# 11. BEST MODEL CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(

    y_test,

    best_predictions,

    labels=all_labels

)

cm_df = pd.DataFrame(

    cm,

    index=class_names,

    columns=class_names

)

cm_df.to_csv(

    OUTPUT_DIR / "best_model_confusion_matrix.csv"

)

disp = ConfusionMatrixDisplay(

    confusion_matrix=cm,

    display_labels=class_names

)

fig, ax = plt.subplots(figsize=(9, 7))

disp.plot(

    cmap="Blues",

    values_format="d",

    xticks_rotation=45,

    ax=ax

)

plt.title(f"{best_model_name} - Confusion Matrix")

plt.tight_layout()

plt.savefig(

    OUTPUT_DIR / "best_model_confusion_matrix.png",

    dpi=300

)

plt.show()

# ============================================================
# 12. SAVE BEST MODEL
# ============================================================

joblib.dump(

    best_model,

    OUTPUT_DIR / "best_classification_model.pkl"

)

# Save best model information

best_config = {

    "best_model": best_model_name,

    "selection_metric": "CV Macro F1",

    "cv_macro_f1": float(
        cv_df.iloc[0]["CV_Macro_F1_Mean"]
    ),

    "features": features,

    "classes": list(class_names),

    "training_records": len(X_train),

    "testing_records": len(X_test)

}

with open(

    OUTPUT_DIR / "best_model_configuration.json",

    "w"

) as file:

    json.dump(

        best_config,

        file,

        indent=4

    )

# ============================================================
# 13. FINAL SUMMARY
# ============================================================

print("\n" + "=" * 70)

print("MULTI-MODEL CLASSIFICATION COMPLETED!")

print("=" * 70)

print("\nSelected Model:", best_model_name)

print(
    "CV Macro F1:",
    round(cv_df.iloc[0]["CV_Macro_F1_Mean"], 4)
)

print("\nSaved Models:")

print("All individual models:", OUTPUT_DIR.resolve())

print("\nBest Model:")

print(
    (OUTPUT_DIR / "best_classification_model.pkl").resolve()
)

print("\nComparison Results:")

print(
    (OUTPUT_DIR / "cross_validation_comparison.csv").resolve()
)
```

```python

# ==========================================================
# FAST GRADIENT BOOSTING - NO HYPERPARAMETER SEARCH
# ==========================================================

import json
import joblib
import numpy as np
import pandas as pd

from pathlib import Path
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# 1. PATHS

CURRENT_DIR = Path.cwd()

if (CURRENT_DIR / "ML_Preprocessing").exists():
    DATA_DIR = CURRENT_DIR / "ML_Preprocessing"
elif CURRENT_DIR.name == "ML_Preprocessing":
    DATA_DIR = CURRENT_DIR
else:
    raise FileNotFoundError("ML_Preprocessing folder not found.")

OUTPUT_DIR = DATA_DIR.parent / "Fast_Model_Optimization"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Loading datasets...")

# 2. LOAD DATA

X_train = pd.read_csv(DATA_DIR / "X_train_rf.csv")
X_test = pd.read_csv(DATA_DIR / "X_test_rf.csv")

y_train_df = pd.read_csv(DATA_DIR / "y_train.csv")
y_test_df = pd.read_csv(DATA_DIR / "y_test.csv")

label_encoder = joblib.load(DATA_DIR / "label_encoder.pkl")

y_train_raw = y_train_df.iloc[:, 0]
y_test_raw = y_test_df.iloc[:, 0]

classes = list(label_encoder.classes_)
class_ids = np.arange(len(classes))

def convert_labels(values):
    values = pd.Series(values).reset_index(drop=True)

    if pd.api.types.is_numeric_dtype(values):
        arr = values.to_numpy()

        if (
            np.isfinite(arr).all()
            and np.all(arr == arr.astype(int))
            and set(arr.astype(int)).issubset(set(class_ids))
        ):
            return arr.astype(int)

    return label_encoder.transform(values.astype(str))

y_train = convert_labels(y_train_raw)
y_test = convert_labels(y_test_raw)

# 3. VALIDATION

assert len(X_train) == len(y_train)
assert len(X_test) == len(y_test)
assert list(X_train.columns) == list(X_test.columns)

if X_train.isnull().any().any() or X_test.isnull().any().any():
    raise ValueError("Missing values found.")

if not np.isfinite(X_train.to_numpy()).all():
    raise ValueError("Invalid training values.")

if not np.isfinite(X_test.to_numpy()).all():
    raise ValueError("Invalid testing values.")

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Features:", len(X_train.columns))
print("Classes:", classes)

# 4. CREATE GRADIENT BOOSTING MODEL

model = GradientBoostingClassifier(
    n_estimators=50,
    learning_rate=0.1,
    max_depth=1,
    min_samples_leaf=4,
    subsample=0.8,
    random_state=42
)

# 5. TRAIN MODEL

print("\nTraining Gradient Boosting Model...")
model.fit(X_train, y_train)

print("Training completed successfully!")

# 6. PREDICTIONS

y_pred = model.predict(X_test)

# 7. EVALUATION

accuracy = accuracy_score(y_test, y_pred)

balanced_acc = balanced_accuracy_score(
    y_test, y_pred
)

precision = precision_score(
    y_test,
    y_pred,
    average="macro",
    labels=class_ids,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_pred,
    average="macro",
    labels=class_ids,
    zero_division=0
)

macro_f1 = f1_score(
    y_test,
    y_pred,
    average="macro",
    labels=class_ids,
    zero_division=0
)

weighted_f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)

print("\nFINAL TEST RESULTS")
print("------------------")
print("Accuracy:", round(accuracy, 4))
print("Balanced Accuracy:", round(balanced_acc, 4))
print("Macro Precision:", round(precision, 4))
print("Macro Recall:", round(recall, 4))
print("Macro F1:", round(macro_f1, 4))
print("Weighted F1:", round(weighted_f1, 4))

# 8. CLASSIFICATION REPORT

report = classification_report(
    y_test,
    y_pred,
    labels=class_ids,
    target_names=classes,
    output_dict=True,
    zero_division=0
)

report_df = pd.DataFrame(report).transpose()

print("\nCLASSIFICATION REPORT")
print(report_df)

report_df.to_csv(
    OUTPUT_DIR / "gb_classification_report.csv"
)

# 9. CONFUSION MATRIX

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=class_ids
)

cm_df = pd.DataFrame(
    cm,
    index=classes,
    columns=classes
)

print("\nCONFUSION MATRIX")
print(cm_df)

cm_df.to_csv(
    OUTPUT_DIR / "gb_confusion_matrix.csv"
)

# 10. FEATURE IMPORTANCE

importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": model.feature_importances_
}).sort_values(
    "Importance",
    ascending=False
)

print("\nFEATURE IMPORTANCE")
print(importance)

importance.to_csv(
    OUTPUT_DIR / "gb_feature_importance.csv",
    index=False
)

# 11. SAVE MODEL

joblib.dump(
    model,
    OUTPUT_DIR / "gradient_boosting_model.pkl"
)

# 12. SAVE CONFIGURATION

configuration = {
    "model": "GradientBoostingClassifier",
    "parameters": model.get_params(),
    "accuracy": float(accuracy),
    "balanced_accuracy": float(balanced_acc),
    "macro_precision": float(precision),
    "macro_recall": float(recall),
    "macro_f1": float(macro_f1),
    "weighted_f1": float(weighted_f1),
    "features": list(X_train.columns),
    "classes": classes
}

with open(
    OUTPUT_DIR / "gb_configuration.json",
    "w"
) as f:
    json.dump(configuration, f, indent=4)

print("\n================================")
print("GRADIENT BOOSTING COMPLETED")
print("================================")
print("Saved files:", OUTPUT_DIR)
```

```python

# ==========================================================
# WATER QUALITY PROJECT
# TEST SET EVALUATION OF ALL SAVED ML MODELS
# ==========================================================

import joblib
import numpy as np
import pandas as pd
import warnings

from pathlib import Path
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

# ----------------------------------------------------------
# 1. PROJECT PATHS
# ----------------------------------------------------------

CURRENT_DIR = Path.cwd()

if (CURRENT_DIR / "ML_Preprocessing").exists():
    DATA_DIR = CURRENT_DIR / "ML_Preprocessing"
elif CURRENT_DIR.name == "ML_Preprocessing":
    DATA_DIR = CURRENT_DIR
else:
    raise FileNotFoundError("ML_Preprocessing folder not found.")

PROJECT_DIR = DATA_DIR.parent

MODEL_DIR = PROJECT_DIR / "Model_Comparison"
FAST_MODEL_DIR = PROJECT_DIR / "Fast_Model_Optimization"

OUTPUT_DIR = PROJECT_DIR / "Final_Model_Test_Evaluation"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Data folder:", DATA_DIR)
print("Model folder:", MODEL_DIR)
print("Fast model folder:", FAST_MODEL_DIR)

# ----------------------------------------------------------
# 2. LOAD TEST DATA
# ----------------------------------------------------------

X_test = pd.read_csv(DATA_DIR / "X_test_rf.csv")

y_test_df = pd.read_csv(DATA_DIR / "y_test.csv")

label_encoder = joblib.load(
    DATA_DIR / "label_encoder.pkl"
)

class_names = list(label_encoder.classes_)
class_ids = np.arange(len(class_names))

y_raw = y_test_df.iloc[:, 0]

# Convert labels safely
def prepare_target(values):

    values = pd.Series(values).reset_index(drop=True)

    if pd.api.types.is_numeric_dtype(values):

        arr = values.to_numpy()

        if (
            np.isfinite(arr).all()
            and np.all(arr == arr.astype(int))
            and set(arr.astype(int)).issubset(set(class_ids))
        ):
            return arr.astype(int)

    return label_encoder.transform(values.astype(str))


y_test = prepare_target(y_raw)

# Validate test data
if len(X_test) != len(y_test):
    raise ValueError("Test features and labels have different lengths.")

if X_test.isnull().any().any():
    raise ValueError("Missing values found in test data.")

if not np.isfinite(X_test.to_numpy()).all():
    raise ValueError("Invalid numerical values in test data.")

print("\nTEST DATA INFORMATION")
print("---------------------")
print("Testing records:", len(X_test))
print("Features:", X_test.shape[1])
print("Classes:", class_names)

# ----------------------------------------------------------
# 3. FIND ALL SAVED MODELS
# ----------------------------------------------------------

model_files = []

for folder in [MODEL_DIR, FAST_MODEL_DIR]:

    if folder.exists():

        for extension in ["*.pkl", "*.joblib"]:

            model_files.extend(folder.glob(extension))

if not model_files:
    raise FileNotFoundError(
        "No saved model files found in the model folders."
    )

print("\nSAVED MODEL FILES FOUND:", len(model_files))

# ----------------------------------------------------------
# 4. LOAD AND EVALUATE EACH MODEL
# ----------------------------------------------------------

results = []
all_reports = {}
all_confusion_matrices = {}

for model_path in model_files:

    model_name = model_path.stem

    print("\n--------------------------------------")
    print("Evaluating:", model_name)
    print("--------------------------------------")

    try:

        model = joblib.load(model_path)

        # Check whether the object supports prediction
        if not hasattr(model, "predict"):
            print("Skipped: Object is not a prediction model.")
            continue

        # Check feature compatibility
        if hasattr(model, "n_features_in_"):
            if model.n_features_in_ != X_test.shape[1]:
                print("Skipped: Feature count mismatch.")
                continue

        # Predict
        y_pred = model.predict(X_test)

        # Evaluate metrics
        accuracy = accuracy_score(y_test, y_pred)

        balanced_accuracy = balanced_accuracy_score(
            y_test, y_pred
        )

        precision = precision_score(
            y_test,
            y_pred,
            labels=class_ids,
            average="macro",
            zero_division=0
        )

        recall = recall_score(
            y_test,
            y_pred,
            labels=class_ids,
            average="macro",
            zero_division=0
        )

        macro_f1 = f1_score(
            y_test,
            y_pred,
            labels=class_ids,
            average="macro",
            zero_division=0
        )

        weighted_f1 = f1_score(
            y_test,
            y_pred,
            labels=class_ids,
            average="weighted",
            zero_division=0
        )

        # Store metrics
        results.append({
            "Model": model_name,
            "Accuracy": accuracy,
            "Balanced_Accuracy": balanced_accuracy,
            "Macro_Precision": precision,
            "Macro_Recall": recall,
            "Macro_F1": macro_f1,
            "Weighted_F1": weighted_f1
        })

        # Classification report
        report = classification_report(
            y_test,
            y_pred,
            labels=class_ids,
            target_names=class_names,
            output_dict=True,
            zero_division=0
        )

        all_reports[model_name] = pd.DataFrame(report).transpose()

        # Confusion matrix
        cm = confusion_matrix(
            y_test,
            y_pred,
            labels=class_ids
        )

        all_confusion_matrices[model_name] = pd.DataFrame(
            cm,
            index=class_names,
            columns=class_names
        )

        print("Accuracy:", round(accuracy, 4))
        print("Balanced Accuracy:", round(balanced_accuracy, 4))
        print("Macro F1:", round(macro_f1, 4))
        print("Weighted F1:", round(weighted_f1, 4))

    except Exception as error:

        print("Could not evaluate this model:", repr(error))

# ----------------------------------------------------------
# 5. CREATE MODEL COMPARISON TABLE
# ----------------------------------------------------------

if not results:
    raise RuntimeError(
        "No saved models could be evaluated successfully."
    )

results_df = pd.DataFrame(results)

results_df = results_df.sort_values(
    by="Macro_F1",
    ascending=False
).reset_index(drop=True)

print("\n==========================================")
print("FINAL TEST SET MODEL COMPARISON")
print("==========================================")

print(results_df.to_string(index=False))

# ----------------------------------------------------------
# 6. SAVE COMPARISON RESULTS
# ----------------------------------------------------------

results_df.to_csv(
    OUTPUT_DIR / "all_models_test_comparison.csv",
    index=False
)

# ----------------------------------------------------------
# 7. SAVE CLASSIFICATION REPORTS
# ----------------------------------------------------------

for model_name, report_df in all_reports.items():

    report_df.to_csv(
        OUTPUT_DIR / f"{model_name}_classification_report.csv"
    )

# ----------------------------------------------------------
# 8. SAVE CONFUSION MATRICES
# ----------------------------------------------------------

for model_name, cm_df in all_confusion_matrices.items():

    cm_df.to_csv(
        OUTPUT_DIR / f"{model_name}_confusion_matrix.csv"
    )

# ----------------------------------------------------------
# 9. SHOW PREVIOUS CROSS-VALIDATION RESULTS
# ----------------------------------------------------------

cv_file = MODEL_DIR / "cross_validation_comparison.csv"

if cv_file.exists():

    cv_results = pd.read_csv(cv_file)

    print("\nPREVIOUS CROSS-VALIDATION RESULTS")
    print("--------------------------------")

    print(cv_results.to_string(index=False))

    cv_results.to_csv(
        OUTPUT_DIR / "previous_cv_comparison_copy.csv",
        index=False
    )

else:

    print("\nPrevious CV comparison file not found.")

# ----------------------------------------------------------
# 10. BEST TEST MACRO F1 MODEL
# ----------------------------------------------------------

best_row = results_df.iloc[0]

print("\n==========================================")
print("BEST MODEL ON THIS TEST SET")
print("==========================================")

print("Model:", best_row["Model"])
print("Accuracy:", round(best_row["Accuracy"], 4))
print("Balanced Accuracy:", round(best_row["Balanced_Accuracy"], 4))
print("Macro F1:", round(best_row["Macro_F1"], 4))

print("\nAll evaluation files saved in:")
print(OUTPUT_DIR)

print("\nTEST EVALUATION COMPLETED SUCCESSFULLY!")
```

```python

# ============================================================
# WATER QUALITY PROJECT
# UNSUPERVISED ANOMALY DETECTION
# ISOLATION FOREST + LOF + ONE-CLASS SVM
# ============================================================

import json
import joblib
import warnings
import numpy as np
import pandas as pd

from pathlib import Path

from sklearn.ensemble import IsolationForest
from sklearn.neighbors import LocalOutlierFactor
from sklearn.svm import OneClassSVM

warnings.filterwarnings("ignore")

# ------------------------------------------------------------
# 1. PROJECT PATHS
# ------------------------------------------------------------

CURRENT_DIR = Path.cwd()

if (CURRENT_DIR / "ML_Preprocessing").exists():
    DATA_DIR = CURRENT_DIR / "ML_Preprocessing"
elif CURRENT_DIR.name == "ML_Preprocessing":
    DATA_DIR = CURRENT_DIR
else:
    raise FileNotFoundError(
        "ML_Preprocessing folder not found."
    )

PROJECT_DIR = DATA_DIR.parent

OUTPUT_DIR = PROJECT_DIR / "Anomaly_Detection"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print("Data folder:", DATA_DIR)
print("Output folder:", OUTPUT_DIR)

# ------------------------------------------------------------
# 2. LOAD PREPROCESSED DATA
# ------------------------------------------------------------

X_train = pd.read_csv(DATA_DIR / "X_train_if.csv")
X_test = pd.read_csv(DATA_DIR / "X_test_if.csv")

train_metadata = pd.read_csv(
    DATA_DIR / "train_metadata.csv"
)

test_metadata = pd.read_csv(
    DATA_DIR / "test_metadata.csv"
)

print("\nDATA LOADED SUCCESSFULLY")

print("Training records:", len(X_train))
print("Testing records:", len(X_test))
print("Features:", X_train.shape[1])

# ------------------------------------------------------------
# 3. VALIDATE DATA
# ------------------------------------------------------------

if list(X_train.columns) != list(X_test.columns):
    raise ValueError(
        "Training and testing feature columns do not match."
    )

if len(train_metadata) != len(X_train):
    raise ValueError(
        "Training metadata row count mismatch."
    )

if len(test_metadata) != len(X_test):
    raise ValueError(
        "Testing metadata row count mismatch."
    )

X_train = X_train.apply(pd.to_numeric, errors="raise")
X_test = X_test.apply(pd.to_numeric, errors="raise")

if X_train.isnull().any().any():
    raise ValueError("Missing values found in training data.")

if X_test.isnull().any().any():
    raise ValueError("Missing values found in testing data.")

if not np.isfinite(X_train.to_numpy()).all():
    raise ValueError("Invalid values found in training data.")

if not np.isfinite(X_test.to_numpy()).all():
    raise ValueError("Invalid values found in testing data.")

print("\nDATA VALIDATION PASSED")

# We use the existing IF-preprocessed files.
# Do not fit a new scaler using the test data.

if (DATA_DIR / "if_scaler.pkl").exists():
    print("Existing Isolation Forest scaler found.")

if (DATA_DIR / "if_imputer.pkl").exists():
    print("Existing Isolation Forest imputer found.")

# ------------------------------------------------------------
# 4. CONFIGURATION
# ------------------------------------------------------------

CONTAMINATION = 0.05

print("\nANOMALY DETECTION CONFIGURATION")
print("--------------------------------")
print("Initial contamination setting:", CONTAMINATION)
print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# ------------------------------------------------------------
# 5. ISOLATION FOREST
# ------------------------------------------------------------

print("\n1. TRAINING ISOLATION FOREST...")

isolation_forest = IsolationForest(
    n_estimators=200,
    contamination=CONTAMINATION,
    max_samples="auto",
    random_state=42,
    n_jobs=2
)

isolation_forest.fit(X_train)

# Prediction:
#  1 = Normal
# -1 = Anomaly

if_predictions = isolation_forest.predict(X_test)

# Higher score means more unusual
if_scores = -isolation_forest.decision_function(X_test)

print("Isolation Forest completed.")

# Save model
joblib.dump(
    isolation_forest,
    OUTPUT_DIR / "isolation_forest_model.pkl"
)

# ------------------------------------------------------------
# 6. LOCAL OUTLIER FACTOR
# ------------------------------------------------------------

print("\n2. TRAINING LOCAL OUTLIER FACTOR...")

n_neighbors = min(20, len(X_train) - 1)

lof_model = LocalOutlierFactor(
    n_neighbors=n_neighbors,
    contamination=CONTAMINATION,
    novelty=True,
    metric="minkowski",
    n_jobs=2
)

lof_model.fit(X_train)

lof_predictions = lof_model.predict(X_test)

# Lower score_samples means more unusual.
# Negate so higher means more unusual.
lof_scores = -lof_model.score_samples(X_test)

print("Local Outlier Factor completed.")

joblib.dump(
    lof_model,
    OUTPUT_DIR / "local_outlier_factor_model.pkl"
)

# ------------------------------------------------------------
# 7. ONE-CLASS SVM
# ------------------------------------------------------------

print("\n3. TRAINING ONE-CLASS SVM...")

one_class_svm = OneClassSVM(
    kernel="rbf",
    gamma="scale",
    nu=CONTAMINATION,
    cache_size=1000,
    max_iter=5000
)

one_class_svm.fit(X_train)

ocsvm_predictions = one_class_svm.predict(X_test)

# Higher score means more unusual
ocsvm_scores = -one_class_svm.decision_function(X_test)

print("One-Class SVM completed.")

joblib.dump(
    one_class_svm,
    OUTPUT_DIR / "one_class_svm_model.pkl"
)

# ------------------------------------------------------------
# 8. CREATE ANOMALY RESULT TABLE
# ------------------------------------------------------------

results = test_metadata.copy()

results["IsolationForest_Prediction"] = if_predictions
results["IsolationForest_Anomaly"] = (
    if_predictions == -1
)
results["IsolationForest_Score"] = if_scores

results["LOF_Prediction"] = lof_predictions
results["LOF_Anomaly"] = (
    lof_predictions == -1
)
results["LOF_Score"] = lof_scores

results["OneClassSVM_Prediction"] = ocsvm_predictions
results["OneClassSVM_Anomaly"] = (
    ocsvm_predictions == -1
)
results["OneClassSVM_Score"] = ocsvm_scores

# ------------------------------------------------------------
# 9. CONSENSUS ANOMALY DETECTION
# ------------------------------------------------------------

anomaly_columns = [
    "IsolationForest_Anomaly",
    "LOF_Anomaly",
    "OneClassSVM_Anomaly"
]

results["Algorithms_Flagging_Anomaly"] = (
    results[anomaly_columns].sum(axis=1)
)

results["Consensus_Anomaly"] = (
    results["Algorithms_Flagging_Anomaly"] >= 2
)

def classify_consensus(count):

    if count == 3:
        return "Flagged by all 3 models"
    elif count == 2:
        return "Flagged by 2 models"
    elif count == 1:
        return "Flagged by 1 model"
    else:
        return "Not flagged"

results["Consensus_Category"] = (
    results["Algorithms_Flagging_Anomaly"]
    .apply(classify_consensus)
)

# ------------------------------------------------------------
# 10. SAVE SAMPLE-LEVEL RESULTS
# ------------------------------------------------------------

results.to_csv(
    OUTPUT_DIR / "anomaly_test_results.csv",
    index=False
)

# ------------------------------------------------------------
# 11. MODEL COMPARISON SUMMARY
# ------------------------------------------------------------

model_predictions = {
    "Isolation Forest": if_predictions,
    "Local Outlier Factor": lof_predictions,
    "One-Class SVM": ocsvm_predictions
}

summary = []

for model_name, predictions in model_predictions.items():

    anomaly_count = int(np.sum(predictions == -1))
    normal_count = int(np.sum(predictions == 1))

    anomaly_rate = anomaly_count / len(predictions) * 100

    summary.append({
        "Model": model_name,
        "Total_Test_Samples": len(predictions),
        "Normal_Samples": normal_count,
        "Anomalous_Samples": anomaly_count,
        "Anomaly_Rate_Percent": round(anomaly_rate, 2)
    })

summary_df = pd.DataFrame(summary)

print("\n========================================")
print("ANOMALY DETECTION MODEL COMPARISON")
print("========================================")

print(summary_df.to_string(index=False))

summary_df.to_csv(
    OUTPUT_DIR / "anomaly_model_comparison.csv",
    index=False
)

# ------------------------------------------------------------
# 12. CONSENSUS SUMMARY
# ------------------------------------------------------------

print("\nCONSENSUS ANOMALY SUMMARY")
print("-------------------------")

consensus_summary = (
    results["Consensus_Category"]
    .value_counts()
    .rename_axis("Category")
    .reset_index(name="Sample_Count")
)

print(consensus_summary.to_string(index=False))

consensus_summary.to_csv(
    OUTPUT_DIR / "consensus_summary.csv",
    index=False
)

print(
    "\nSamples flagged by at least 2 algorithms:",
    int(results["Consensus_Anomaly"].sum())
)

# ------------------------------------------------------------
# 13. REVIEW GROUP ANALYSIS, IF AVAILABLE
# ------------------------------------------------------------

if "review_group" in results.columns:

    print("\nREVIEW GROUP VS ANOMALY DETECTION")
    print("--------------------------------")

    review_summary = results.groupby(
        "review_group",
        dropna=False
    ).agg(
        Total_Samples=("Consensus_Anomaly", "size"),
        Consensus_Anomalies=("Consensus_Anomaly", "sum")
    ).reset_index()

    review_summary["Anomaly_Percentage"] = (
        review_summary["Consensus_Anomalies"]
        / review_summary["Total_Samples"] * 100
    ).round(2)

    print(review_summary.to_string(index=False))

    review_summary.to_csv(
        OUTPUT_DIR / "review_group_anomaly_analysis.csv",
        index=False
    )

# ------------------------------------------------------------
# 14. SAVE CONFIGURATION
# ------------------------------------------------------------

configuration = {
    "models": [
        "Isolation Forest",
        "Local Outlier Factor",
        "One-Class SVM"
    ],
    "contamination_setting": CONTAMINATION,
    "training_samples": len(X_train),
    "testing_samples": len(X_test),
    "number_of_features": X_train.shape[1],
    "features": list(X_train.columns),
    "isolation_forest_estimators": 200,
    "lof_neighbors": n_neighbors,
    "one_class_svm_kernel": "rbf",
    "one_class_svm_gamma": "scale",
    "consensus_rule": "At least 2 of 3 models flag the sample"
}

with open(
    OUTPUT_DIR / "anomaly_configuration.json",
    "w"
) as file:
    json.dump(configuration, file, indent=4)

# ------------------------------------------------------------
# 15. COMPLETION
# ------------------------------------------------------------

print("\n========================================")
print("ANOMALY DETECTION COMPLETED")
print("========================================")

print("\nSAVED FILES:")
print("1. isolation_forest_model.pkl")
print("2. local_outlier_factor_model.pkl")
print("3. one_class_svm_model.pkl")
print("4. anomaly_test_results.csv")
print("5. anomaly_model_comparison.csv")
print("6. consensus_summary.csv")
print("7. anomaly_configuration.json")

print("\nOutput folder:", OUTPUT_DIR)
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

BASE_DIR = Path(
    r"C:\Water Quality and Anomolgy detetction\cleaned_files"
)

PREP_DIR = BASE_DIR / "ML_Preprocessing"
OUT_DIR = BASE_DIR / "Anomaly_Detection"

# Load transformed training and test features
X_train = pd.read_csv(PREP_DIR / "X_train_if.csv")
X_test = pd.read_csv(PREP_DIR / "X_test_if.csv")

# Load anomaly results
results = pd.read_csv(OUT_DIR / "anomaly_test_results.csv")

# Reset indexes to preserve row-wise alignment
X_train = X_train.reset_index(drop=True)
X_test = X_test.reset_index(drop=True)
results = results.reset_index(drop=True)

assert len(X_test) == len(results), "Test rows do not match anomaly results"
assert list(X_train.columns) == list(X_test.columns), "Feature mismatch"

# Select consensus anomalies
flagged = results[
    results["Consensus_Anomaly"].astype(str).str.lower().isin(
        ["true", "1"]
    )
].copy()

flagged_positions = flagged.index.tolist()

# Calculate training median and IQR
train_median = X_train.median()
train_iqr = X_train.quantile(0.75) - X_train.quantile(0.25)

# Avoid division by zero
train_iqr = train_iqr.replace(0, np.nan)

# Robust deviation from training distribution
deviations = (
    X_test.iloc[flagged_positions] - train_median
).div(train_iqr)

deviations = deviations.replace([np.inf, -np.inf], np.nan)

# Top unusual parameters for every flagged sample
detail_rows = []

for position in flagged_positions:
    row_deviation = deviations.loc[position].abs().sort_values(
        ascending=False
    )

    for feature in row_deviation.head(5).index:
        detail_rows.append({
            "Test_Position": position,
            "Station_ID": results.loc[position, "station_id"],
            "Station_Name": results.loc[position, "station_name"],
            "Year": results.loc[position, "year"],
            "Review_Group": results.loc[position, "review_group"],
            "Feature": feature,
            "Signed_Robust_Deviation": deviations.loc[position, feature],
            "Absolute_Deviation": abs(deviations.loc[position, feature])
        })

detail_df = pd.DataFrame(detail_rows)

# Count how often each feature appears in the top 5
feature_summary = (
    detail_df.groupby("Feature")
    .agg(
        Times_In_Top_5=("Feature", "count"),
        Average_Absolute_Deviation=("Absolute_Deviation", "mean")
    )
    .sort_values("Times_In_Top_5", ascending=False)
    .reset_index()
)

# Save results
detail_df.to_csv(
    OUT_DIR / "anomaly_feature_investigation.csv",
    index=False
)

feature_summary.to_csv(
    OUT_DIR / "anomaly_feature_summary.csv",
    index=False
)

print("TOP CHEMICAL PARAMETERS ASSOCIATED WITH ANOMALIES")
display(feature_summary)

print("\nTOP 5 PARAMETERS PER FLAGGED SAMPLE")
display(detail_df)

print("\nSPLIT METADATA CHECK")
print(results["suggested_split"].value_counts(dropna=False))

print("\nSaved:")
print("anomaly_feature_investigation.csv")
print("anomaly_feature_summary.csv")
```

```python

import pandas as pd
import numpy as np
from pathlib import Path

BASE = Path(
    r"C:\Water Quality and Anomolgy detetction\cleaned_files"
)

PREP = BASE / "ML_Preprocessing"
OUT = BASE / "Anomaly_Detection"

# Load files
raw = pd.read_csv(BASE / "nwmp_model_ready.csv")
train_meta = pd.read_csv(PREP / "train_metadata.csv")
test_meta = pd.read_csv(PREP / "test_metadata.csv")
results = pd.read_csv(OUT / "anomaly_test_results.csv")

# Original chemical parameters only
features = [
    "pH", "DO", "BOD", "Cond", "Nitrate", "COD",
    "TotalColi", "FecalColi", "Turbidity", "TotAlk",
    "Chloride", "Sulphate", "Hardness", "TDS",
    "TSS", "Fluoride"
]

keys = ["station_id", "year"]

# Validate station-year matching
assert not raw.duplicated(keys).any(), "Duplicate keys in raw data"
assert not train_meta.duplicated(keys).any(), "Duplicate training keys"
assert not test_meta.duplicated(keys).any(), "Duplicate test keys"
assert len(test_meta) == len(results), "Test row count mismatch"

# Match original measurements to actual training/test records
train_data = train_meta[keys].merge(
    raw[keys + features],
    on=keys,
    how="left",
    validate="one_to_one",
    indicator=True
)

test_data = test_meta[keys].merge(
    raw[keys + features],
    on=keys,
    how="left",
    validate="one_to_one",
    indicator=True
)

assert (train_data["_merge"] == "both").all(), "Training match failed"
assert (test_data["_merge"] == "both").all(), "Test match failed"

train_data.drop(columns="_merge", inplace=True)
test_data.drop(columns="_merge", inplace=True)

# Verify numeric data
for col in features:
    train_data[col] = pd.to_numeric(
        train_data[col], errors="coerce"
    )
    test_data[col] = pd.to_numeric(
        test_data[col], errors="coerce"
    )

print("TRAIN RECORDS:", len(train_data))
print("TEST RECORDS:", len(test_data))
print("\nMissing training measurements:")
print(train_data[features].isna().sum())

# Calculate reference statistics from training records only
train_median = train_data[features].median()
train_q1 = train_data[features].quantile(0.25)
train_q3 = train_data[features].quantile(0.75)
train_iqr = (train_q3 - train_q1).replace(0, np.nan)

# Join anomaly results to original test measurements
test_data = pd.concat(
    [test_data.reset_index(drop=True),
     results.reset_index(drop=True)],
    axis=1
)

# Select consensus anomalies
flagged = test_data[
    test_data["Consensus_Anomaly"].astype(str).str.lower()
    .isin(["true", "1"])
].copy()

# Calculate robust deviations using original measurements
detail_rows = []

for _, row in flagged.iterrows():
    deviations = (
        (row[features].astype(float) - train_median) / train_iqr
    ).replace([np.inf, -np.inf], np.nan).dropna()

    top_features = deviations.abs().sort_values(
        ascending=False
    ).head(5)

    for feature in top_features.index:
        detail_rows.append({
            "station_id": row["station_id"],
            "station_name": row["station_name"],
            "year": row["year"],
            "review_group": row["review_group"],
            "Consensus_Category": row["Consensus_Category"],
            "Chemical_Parameter": feature,
            "Original_Measurement": row[feature],
            "Training_Median": train_median[feature],
            "Training_Q1": train_q1[feature],
            "Training_Q3": train_q3[feature],
            "Robust_Deviation_IQR": deviations[feature]
        })

detail = pd.DataFrame(detail_rows)

# Parameter frequency summary
summary = (
    detail.groupby("Chemical_Parameter")
    .agg(
        Times_In_Top_5=("Chemical_Parameter", "count"),
        Mean_Absolute_Deviation=(
            "Robust_Deviation_IQR",
            lambda x: x.abs().mean()
        )
    )
    .sort_values(
        "Times_In_Top_5", ascending=False
    )
    .reset_index()
)

# Save results
detail.to_csv(
    OUT / "original_unit_anomaly_investigation.csv",
    index=False
)

summary.to_csv(
    OUT / "original_unit_anomaly_summary.csv",
    index=False
)

flagged.to_csv(
    OUT / "consensus_anomalies_with_original_values.csv",
    index=False
)

print("\nORIGINAL MEASUREMENTS: TOP PARAMETERS")
display(summary)

print("\nTOP 5 PARAMETERS PER ANOMALOUS RECORD")
display(detail)

print("\nORIGINAL VALUES OF ALL CONSENSUS ANOMALIES")
display(flagged[["station_id", "station_name", "year",
                 "review_group"] + features])

print("\nFILES SAVED IN:", OUT)
```

```python

import pandas as pd
import numpy as np
import joblib
from pathlib import Path
from sklearn.metrics import f1_score

BASE = Path(r"C:\Water Quality and Anomolgy detetction\cleaned_files")
PREP = BASE / "ML_Preprocessing"
OUT = BASE / "Model_Explainability"
OUT.mkdir(exist_ok=True)

print("Loading files...")

model = joblib.load(
    BASE / "Model_Comparison" / "best_classification_model.pkl"
)

X_test = pd.read_csv(PREP / "X_test_rf.csv")
y_test = pd.read_csv(PREP / "y_test.csv").squeeze()

print("Calculating original Macro F1...")

baseline = f1_score(
    y_test,
    model.predict(X_test),
    average="macro",
    zero_division=0
)

print("Baseline Macro F1:", round(baseline, 4))

results = []
rng = np.random.default_rng(42)

for i, column in enumerate(X_test.columns, start=1):

    print(f"Processing {i}/{len(X_test.columns)}: {column}")

    decreases = []

    for repeat in range(3):
        X_shuffled = X_test.copy()

        X_shuffled[column] = rng.permutation(
            X_shuffled[column].to_numpy()
        )

        predictions = model.predict(X_shuffled)

        shuffled_f1 = f1_score(
            y_test,
            predictions,
            average="macro",
            zero_division=0
        )

        decreases.append(baseline - shuffled_f1)

    results.append({
        "Chemical_Parameter": column,
        "Mean_F1_Decrease": np.mean(decreases),
        "Std_F1_Decrease": np.std(decreases)
    })

result_df = pd.DataFrame(results)

result_df = result_df.sort_values(
    "Mean_F1_Decrease",
    ascending=False
)

print("\nPERMUTATION IMPORTANCE RESULTS")
display(result_df)

result_df.to_csv(
    OUT / "permutation_importance_lightweight.csv",
    index=False
)

print("\nCompleted successfully!")
print("Saved at:", OUT)
```

```python

import pandas as pd
import numpy as np
import joblib
import shap
import matplotlib.pyplot as plt

from pathlib import Path
from sklearn.preprocessing import LabelEncoder

BASE = Path(r"C:\Water Quality and Anomolgy detetction\cleaned_files")
PREP = BASE / "ML_Preprocessing"
OUT = BASE / "Model_Explainability"
OUT.mkdir(exist_ok=True)

print("Loading model and datasets...")

model = joblib.load(
    BASE / "Model_Comparison" / "best_classification_model.pkl"
)

X_train = pd.read_csv(PREP / "X_train_rf.csv")
X_test = pd.read_csv(PREP / "X_test_rf.csv")

label_encoder = joblib.load(
    PREP / "label_encoder.pkl"
)

class_names = list(label_encoder.classes_)

# Small background and explanation samples
background = X_train.sample(
    n=min(30, len(X_train)),
    random_state=42
)

X_sample = X_test.sample(
    n=min(30, len(X_test)),
    random_state=42
)

print("Background samples:", len(background))
print("Samples to explain:", len(X_sample))

# Explain predicted class probabilities
def predict_probability(data):
    data = pd.DataFrame(
        data,
        columns=X_train.columns
    )
    return model.predict_proba(data)

print("Creating KernelExplainer...")

explainer = shap.KernelExplainer(
    predict_probability,
    background
)

print("Calculating SHAP values...")

shap_values = explainer.shap_values(
    X_sample,
    nsamples=100
)

print("SHAP calculation completed!")

# Normalize output into one array per class
if isinstance(shap_values, list):
    values_by_class = [
        np.asarray(v) for v in shap_values
    ]
else:
    values = np.asarray(shap_values)

    if values.ndim == 3:
        if values.shape[0] == len(X_sample):
            values_by_class = [
                values[:, :, i]
                for i in range(values.shape[2])
            ]
        elif values.shape[1] == len(X_sample):
            values_by_class = [
                values[i, :, :]
                for i in range(values.shape[0])
            ]
        else:
            raise ValueError(
                f"Unexpected SHAP shape: {values.shape}"
            )
    else:
        raise ValueError(
            f"Unexpected SHAP shape: {values.shape}"
        )

# Overall mean absolute SHAP importance
overall = np.mean(
    np.stack([
        np.abs(v) for v in values_by_class
    ]),
    axis=(0, 1)
)

importance_df = pd.DataFrame({
    "Chemical_Parameter": X_sample.columns,
    "Mean_Absolute_SHAP": overall
}).sort_values(
    "Mean_Absolute_SHAP",
    ascending=False
)

print("\nOVERALL SHAP IMPORTANCE")
display(importance_df)

importance_df.to_csv(
    OUT / "kernel_shap_importance.csv",
    index=False
)

# Class-wise summary plots
for i, class_name in enumerate(class_names):

    print("Plotting:", class_name)

    plt.figure()

    shap.summary_plot(
        values_by_class[i],
        X_sample,
        show=False,
        max_display=16
    )

    plt.title(f"SHAP Explanation - {class_name}")
    plt.tight_layout()

    filename = str(class_name).replace(" ", "_")

    plt.savefig(
        OUT / f"kernel_shap_{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()

print("\nALL SHAP FILES SAVED IN:", OUT)
```

```python

import pandas as pd
import numpy as np
import joblib
import shap

from pathlib import Path

BASE = Path(r"C:\Water Quality and Anomolgy detetction\cleaned_files")
PREP = BASE / "ML_Preprocessing"
OUT = BASE / "Model_Explainability"
OUT.mkdir(exist_ok=True)

# Load trained model and data
model = joblib.load(
    BASE / "Model_Comparison" / "best_classification_model.pkl"
)

X_train = pd.read_csv(PREP / "X_train_rf.csv")
X_test = pd.read_csv(PREP / "X_test_rf.csv")

encoder = joblib.load(
    PREP / "label_encoder.pkl"
)

# Select one sample
sample = X_test.iloc[[0]].copy()

# Create background data
background = X_train.sample(
    n=30,
    random_state=42
)

# Prediction
probabilities = model.predict_proba(sample)[0]
predicted_index = np.argmax(probabilities)
predicted_class = encoder.inverse_transform(
    [predicted_index]
)[0]

print("PREDICTED CLASS:", predicted_class)
print("CLASS PROBABILITIES:")

for name, probability in zip(
    encoder.classes_, probabilities
):
    print(f"{name}: {probability:.4f}")

# SHAP explanation
explainer = shap.KernelExplainer(
    model.predict_proba,
    background
)

print("\nCalculating individual SHAP values...")

raw = explainer.shap_values(
    sample,
    nsamples=100
)

# Handle SHAP output formats
if isinstance(raw, list):
    class_values = [np.asarray(v) for v in raw]
else:
    arr = np.asarray(raw)

    if arr.ndim == 3 and arr.shape[0] == 1:
        class_values = [
            arr[:, :, i] for i in range(arr.shape[2])
        ]
    elif arr.ndim == 3 and arr.shape[1] == 1:
        class_values = [
            arr[i, :, :] for i in range(arr.shape[0])
        ]
    else:
        raise ValueError(
            f"Unexpected SHAP output shape: {arr.shape}"
        )

# Explain the predicted class
values = class_values[predicted_index].reshape(-1)

explanation = pd.DataFrame({
    "Parameter": sample.columns,
    "Measured_Value": sample.iloc[0].values,
    "SHAP_Contribution": values
})

explanation["Effect"] = np.where(
    explanation["SHAP_Contribution"] > 0,
    "Pushes towards predicted class",
    "Pushes away from predicted class"
)

explanation["Absolute_Contribution"] = (
    explanation["SHAP_Contribution"].abs()
)

explanation = explanation.sort_values(
    "Absolute_Contribution",
    ascending=False
)

print("\nINDIVIDUAL SAMPLE EXPLANATION")
display(explanation)

explanation.to_csv(
    OUT / "individual_sample_explanation.csv",
    index=False
)

print("\nSaved individual explanation successfully!")
```

```python

import math
import pandas as pd
import numpy as np

# BIS IS 10500:2012 reference screening limits
BIS_LIMITS = {
    "pH": {
        "unit": "pH",
        "acceptable_min": 6.5,
        "acceptable_max": 8.5,
        "permissible_min": 6.5,
        "permissible_max": 8.5
    },
    "Turbidity": {
        "unit": "NTU",
        "acceptable": 1,
        "permissible": 5
    },
    "TDS": {
        "unit": "mg/L",
        "acceptable": 500,
        "permissible": 2000
    },
    "Chloride": {
        "unit": "mg/L",
        "acceptable": 250,
        "permissible": 1000
    },
    "Fluoride": {
        "unit": "mg/L",
        "acceptable": 1.0,
        "permissible": 1.5
    },
    "Nitrate": {
        "unit": "mg/L",
        "acceptable": 45,
        "permissible": 45
    },
    "Sulphate": {
        "unit": "mg/L",
        "acceptable": 200,
        "permissible": 400
    },
    "Hardness": {
        "unit": "mg/L as CaCO3",
        "acceptable": 200,
        "permissible": 600
    },
    "TotAlk": {
        "unit": "mg/L as CaCO3",
        "acceptable": 200,
        "permissible": 600
    }
}

print("Drinking-water assessment rules loaded!")
print("Parameters configured:", len(BIS_LIMITS))
```

```python

def assess_drinking_suitability(
    measurements,
    microbiological_verified=False
):

    results = {}
    missing = []
    exceeded = []
    permissible = []

    for parameter, limits in BIS_LIMITS.items():

        value = measurements.get(parameter)

        if value is None:
            missing.append(parameter)

            results[parameter] = {
                "value": None,
                "status": "MISSING",
                "unit": limits["unit"]
            }
            continue

        try:
            value = float(value)

            if not math.isfinite(value):
                raise ValueError()

            if parameter != "pH" and value < 0:
                raise ValueError()

        except (ValueError, TypeError):
            raise ValueError(
                f"Invalid value for {parameter}"
            )

        if parameter == "pH":

            if not 6.5 <= value <= 8.5:
                status = "ABOVE_LIMIT"
                exceeded.append(parameter)
            else:
                status = "ACCEPTABLE"

        else:

            if value <= limits["acceptable"]:
                status = "ACCEPTABLE"

            elif value <= limits["permissible"]:
                status = "PERMISSIBLE"
                permissible.append(parameter)

            else:
                status = "ABOVE_LIMIT"
                exceeded.append(parameter)

        results[parameter] = {
            "value": value,
            "unit": limits["unit"],
            "status": status
        }

    if exceeded:
        outcome = "NOT_SUITABLE"

    elif missing or not microbiological_verified:
        outcome = "INSUFFICIENT_DATA"

    elif permissible:
        outcome = "WITHIN_PERMISSIBLE_LIMITS"

    else:
        outcome = "MEETS_ASSESSED_LIMITS"

    return {
        "drinking_suitability": outcome,
        "parameter_results": results,
        "missing_parameters": missing,
        "exceeded_parameters": exceeded,
        "permissible_parameters": permissible,
        "microbiological_verified": microbiological_verified
    }

print("Assessment function created successfully!")
```

```python

# Use original-unit training measurements as reference
train_reference = train_raw[FEATURES].apply(
    pd.to_numeric, errors="coerce"
)

def check_sample_quality(sample):

    findings = []
    missing = []
    invalid = []
    extreme = []

    for feature in FEATURES:

        value = sample.get(feature)

        if value is None or pd.isna(value):
            missing.append(feature)
            continue

        try:
            value = float(value)

            if not np.isfinite(value):
                raise ValueError()

        except (TypeError, ValueError):
            invalid.append(feature)
            findings.append(
                f"{feature}: invalid or non-numeric value"
            )
            continue

        if feature == "pH":

            if value < 0 or value > 14:
                invalid.append(feature)
                findings.append(
                    f"{feature}: outside physical pH scale"
                )

        elif value < 0:

            invalid.append(feature)
            findings.append(
                f"{feature}: negative measurement"
            )

        # Compare with training distribution using IQR
        reference = train_reference[feature].dropna()

        if len(reference) >= 10:

            q1 = reference.quantile(0.25)
            q3 = reference.quantile(0.75)
            iqr = q3 - q1

            if iqr > 0:

                lower = q1 - 1.5 * iqr
                upper = q3 + 1.5 * iqr

                if value < lower or value > upper:
                    extreme.append(feature)

                    findings.append(
                        f"{feature}: statistically extreme "
                        f"relative to training data"
                    )

    # Suspicious relationship, not automatic error
    bod = sample.get("BOD")
    cod = sample.get("COD")

    if (
        bod is not None and cod is not None
        and pd.notna(bod) and pd.notna(cod)
        and float(bod) > float(cod)
    ):
        findings.append(
            "BOD exceeds COD: review measurement, units "
            "and laboratory records"
        )

    if invalid:
        status = "INVALID_INPUT"

    elif missing:
        status = "MISSING_DATA"

    elif findings:
        status = "REQUIRES_REVIEW"

    else:
        status = "PASS"

    return {
        "status": status,
        "missing_parameters": missing,
        "invalid_parameters": invalid,
        "extreme_parameters": extreme,
        "findings": findings
    }

print("Smart Data Quality Checker created!")
```

```python

def predict_with_uncertainty(sample):

    sample_df = pd.DataFrame(
        [sample],
        columns=FEATURES
    )

    if sample_df.isnull().any().any():
        raise ValueError(
            "Complete all 16 measurements before prediction."
        )

    sample_df = sample_df.astype(float)

    if not np.isfinite(sample_df.to_numpy()).all():
        raise ValueError("Measurements must be finite.")

    # Reuse saved training imputer
    transformed = rf_imputer.transform(sample_df)

    transformed = pd.DataFrame(
        transformed,
        columns=FEATURES
    )

    probabilities = classification_model.predict_proba(
        transformed
    )[0]

    model_classes = np.asarray(
        classification_model.classes_,
        dtype=int
    )

    class_names = label_encoder.inverse_transform(
        model_classes
    )

    order = np.argsort(probabilities)[::-1]

    predicted_class = class_names[order[0]]
    top_probability = probabilities[order[0]]
    second_probability = probabilities[order[1]]

    margin = top_probability - second_probability

    # Normalized entropy: 0 = concentrated, 1 = uniform
    safe_probabilities = np.clip(
        probabilities, 1e-12, 1
    )

    entropy = -np.sum(
        safe_probabilities * np.log(safe_probabilities)
    )

    normalized_entropy = entropy / np.log(len(probabilities))

    # Heuristic review thresholds, not calibrated confidence limits
    review_required = (
        top_probability < 0.60 or margin < 0.15
    )

    return {
        "predicted_class": str(predicted_class),
        "class_probabilities": {
            str(name): float(prob)
            for name, prob in zip(class_names, probabilities)
        },
        "top_probability": float(top_probability),
        "second_probability": float(second_probability),
        "probability_margin": float(margin),
        "normalized_entropy": float(normalized_entropy),
        "review_required": bool(review_required),
        "review_status": (
            "MANUAL_REVIEW_RECOMMENDED"
            if review_required else "NO_REVIEW_FLAG"
        )
    }

print("Prediction uncertainty module created!")
```

```python

train_median = train_reference.median()

train_iqr = (
    train_reference.quantile(0.75)
    - train_reference.quantile(0.25)
).replace(0, np.nan)

anomaly_explanations = []

for i in range(len(test_raw)):

    sample = test_raw.iloc[i][FEATURES]

    
    # Convert all values to numeric
    sample_numeric = pd.to_numeric(sample, errors="coerce")
    median_numeric = pd.to_numeric(train_median, errors="coerce")
    iqr_numeric = pd.to_numeric(train_iqr, errors="coerce")
    
    # Avoid division by zero
    iqr_numeric = iqr_numeric.replace(0, np.nan)
    
    # Calculate normalized deviation scores
    deviation_scores = (
        (sample_numeric - median_numeric).abs() / iqr_numeric
    )
    
    # Ensure numeric dtype and remove invalid values
    deviation_scores = pd.to_numeric(
        deviation_scores, errors="coerce"
    ).replace([np.inf, -np.inf], np.nan)
    
    # Select top 5 deviations
    top_features = deviation_scores.dropna().nlargest(5)
    
    print("Top 5 chemical deviations:")
    display(top_features)


    details = []

    for feature, score in top_features.items():

        value = sample[feature]
        median = train_median[feature]

        direction = (
            "higher" if value > median else "lower"
        )

        details.append(
            f"{feature}: {direction} than training median "
            f"({value:.3f} vs {median:.3f}; "
            f"{score:.2f} IQR units)"
        )

    anomaly_explanations.append({
        "test_index": i,
        "station_id": test_raw.iloc[i]["station_id"],
        "year": test_raw.iloc[i]["year"],
        "top_deviating_parameters": ", ".join(
            top_features.index.tolist()
        ),
        "deviation_explanation": " | ".join(details)
    })

anomaly_explanations = pd.DataFrame(
    anomaly_explanations
)

# Combine anomaly flags and deviation explanations
anomaly_results["test_index"] = np.arange(
    len(anomaly_results)
)

anomaly_results = anomaly_results.merge(
    anomaly_explanations,
    on=["test_index", "station_id", "year"],
    how="left",
    validate="one_to_one"
)

print("EXPLAINABLE ANOMALY RESULTS")

display(
    anomaly_results[
        anomaly_results["consensus_votes"] >= 2
    ][
        ["station_id", "year",
         "consensus_votes",
         "top_deviating_parameters",
         "deviation_explanation"]
    ]
)

anomaly_results.to_csv(
    NEW_DIR / "explainable_anomaly_results.csv",
    index=False
)

anomaly_explanations.to_csv(
    NEW_DIR / "chemical_anomaly_explanations.csv",
    index=False
)

print("Anomaly explanations saved!")
```

```python

BIS_LIMITS = {
    "pH": {
        "acceptable_min": 6.5,
        "acceptable_max": 8.5,
        "permissible_max": 8.5,
        "unit": ""
    },
    "Turbidity": {
        "acceptable_max": 1,
        "permissible_max": 5,
        "unit": "NTU"
    },
    "TDS": {
        "acceptable_max": 500,
        "permissible_max": 2000,
        "unit": "mg/L"
    },
    "Chloride": {
        "acceptable_max": 250,
        "permissible_max": 1000,
        "unit": "mg/L"
    },
    "Fluoride": {
        "acceptable_max": 1.0,
        "permissible_max": 1.5,
        "unit": "mg/L"
    },
    "Nitrate": {
        "acceptable_max": 45,
        "permissible_max": 45,
        "unit": "mg/L"
    },
    "Sulphate": {
        "acceptable_max": 200,
        "permissible_max": 400,
        "unit": "mg/L"
    },
    "Hardness": {
        "acceptable_max": 200,
        "permissible_max": 600,
        "unit": "mg/L as CaCO3"
    },
    "TotAlk": {
        "acceptable_max": 200,
        "permissible_max": 600,
        "unit": "mg/L as CaCO3"
    }
}


def assess_drinking_suitability(
    measurements,
    microbiological_verified=False
):
    failed_parameters = []
    warnings = []
    invalid_parameters = []
    missing_parameters = []

    for parameter, limits in BIS_LIMITS.items():

        value = measurements.get(parameter)

        if value is None:
            missing_parameters.append(parameter)
            continue

        try:
            value = float(value)
        except (TypeError, ValueError):
            invalid_parameters.append(parameter)
            continue

        if not np.isfinite(value) or value < 0:
            invalid_parameters.append(parameter)
            continue

        # pH has a range, not just an upper limit
        if parameter == "pH":
            if (
                value < limits["acceptable_min"]
                or value > limits["acceptable_max"]
            ):
                failed_parameters.append({
                    "parameter": parameter,
                    "value": value,
                    "reason": "Outside pH range 6.5–8.5"
                })

        elif value > limits["permissible_max"]:
            failed_parameters.append({
                "parameter": parameter,
                "value": value,
                "permissible_limit": limits["permissible_max"],
                "unit": limits["unit"]
            })

        elif value > limits["acceptable_max"]:
            warnings.append({
                "parameter": parameter,
                "value": value,
                "acceptable_limit": limits["acceptable_max"],
                "permissible_limit": limits["permissible_max"],
                "unit": limits["unit"]
            })

    # A failed parameter takes priority over incomplete verification
    if failed_parameters:
        status = "NOT_SUITABLE"
        message = (
            "One or more assessed parameters exceed "
            "the applicable permissible limit."
        )

    elif invalid_parameters or missing_parameters:
        status = "INSUFFICIENT_DATA"
        message = (
            "Some required drinking-water measurements "
            "are missing or invalid."
        )

    elif not microbiological_verified:
        status = "INSUFFICIENT_DATA"
        message = (
            "Chemical screening completed, but "
            "microbiological laboratory verification is required."
        )

    elif warnings:
        status = "WITHIN_PERMISSIBLE_LIMITS"
        message = (
            "Some parameters exceed acceptable limits "
            "but remain within permissible limits."
        )

    else:
        status = "MEETS_ASSESSED_LIMITS"
        message = (
            "All assessed chemical parameters meet "
            "the selected acceptable limits."
        )

    return {
        "status": status,
        "message": message,
        "parameters_checked": (
            len(BIS_LIMITS) - len(missing_parameters)
            - len(invalid_parameters)
        ),
        "failed_parameters": failed_parameters,
        "warnings": warnings,
        "missing_parameters": missing_parameters,
        "invalid_parameters": invalid_parameters,
        "microbiological_verified": bool(
            microbiological_verified
        ),
        "disclaimer": (
            "Preliminary chemical screening only. "
            "This does not certify drinking-water safety "
            "or complete BIS compliance."
        )
    }


print("Drinking-water suitability module ready!")
```

```python

@app.post("/predict")
def predict_water_quality(sample: WaterSample):

    try:
        sample_dict = sample.dict()

        microbiological_verified = sample_dict.pop(
            "microbiological_verified"
        )

        sample_df = pd.DataFrame(
            [[sample_dict[f] for f in FEATURES]],
            columns=FEATURES
        )

        # 1. Water quality classification
        X_classification = rf_imputer.transform(sample_df)

        predicted_encoded = classification_model.predict(
            X_classification
        )[0]

        predicted_class = label_encoder.inverse_transform(
            np.array([int(predicted_encoded)])
        )[0]

        probabilities = classification_model.predict_proba(
            X_classification
        )[0]

        class_probabilities = {}

        for class_id, probability in zip(
            classification_model.classes_, probabilities
        ):
            class_name = label_encoder.inverse_transform(
                np.array([int(class_id)])
            )[0]

            class_probabilities[str(class_name)] = round(
                float(probability), 4
            )

        top_probability = float(max(probabilities))
        sorted_probabilities = sorted(probabilities, reverse=True)

        margin = (
            sorted_probabilities[0] - sorted_probabilities[1]
            if len(sorted_probabilities) > 1 else 1.0
        )

        # 2. Drinking water suitability
        drinking_result = assess_drinking_suitability(
            sample_dict,
            microbiological_verified
        )

        # 3. Anomaly detection
        X_anomaly = if_imputer.transform(sample_df)
        X_anomaly = if_scaler.transform(X_anomaly)

        if_flag = isolation_forest.predict(X_anomaly)[0] == -1
        lof_flag = lof_model.predict(X_anomaly)[0] == -1
        svm_flag = ocsvm_model.predict(X_anomaly)[0] == -1

        votes = sum([if_flag, lof_flag, svm_flag])

        if votes >= 2:
            anomaly_status = "ANOMALY"
        elif votes == 1:
            anomaly_status = "REVIEW_REQUIRED"
        else:
            anomaly_status = "NORMAL"

        # 4. Chemical deviation explanation
        sample_series = pd.Series(sample_dict)[FEATURES]

        deviations = (
            (sample_series - train_median).abs() / train_iqr
        )

        deviations = pd.to_numeric(
            deviations, errors="coerce"
        ).replace([np.inf, -np.inf], np.nan)

        top_deviations = deviations.dropna().nlargest(5)

        explanation = [
            {
                "parameter": str(name),
                "deviation_score": round(float(value), 4)
            }
            for name, value in top_deviations.items()
        ]

        # 5. Data quality warnings
        warnings = []

        negative = [
            f for f in FEATURES if sample_dict[f] < 0
        ]

        if negative:
            warnings.append(
                "Negative measurements: " + ", ".join(negative)
            )

        if sample_dict["BOD"] > sample_dict["COD"]:
            warnings.append("BOD is greater than COD; verify input.")

        # 6. Combined response
        return {
            "success": True,

            "water_quality": {
                "predicted_class": str(predicted_class),
                "confidence": round(top_probability, 4),
                "class_probabilities": class_probabilities,
                "review_required": bool(
                    top_probability < 0.60 or margin < 0.15
                )
            },

            "drinking_suitability": drinking_result,

            "anomaly_detection": {
                "status": anomaly_status,
                "consensus_votes": int(votes),
                "models": {
                    "isolation_forest": (
                        "ANOMALY" if if_flag else "NORMAL"
                    ),
                    "local_outlier_factor": (
                        "ANOMALY" if lof_flag else "NORMAL"
                    ),
                    "one_class_svm": (
                        "ANOMALY" if svm_flag else "NORMAL"
                    )
                },
                "chemical_deviation_explanation": explanation
            },

            "data_quality": {
                "status": "REQUIRES_REVIEW" if warnings else "PASS",
                "warnings": warnings
            },

            "mongodb_saved": False
        }

    except Exception as error:
        raise HTTPException(
            status_code=500,
            detail=str(error)
        )


print("Prediction endpoint registered successfully!")
```
