```python
"""
NWMP water-quality project: supervised models + anomaly detection + importance.

Usage:  python nwmp_models.py nwmp_model_ready.csv [out_dir]

EDIT THE CONFIG BLOCK if your column names differ (print df.columns to check).
Needs: pandas numpy scikit-learn
"""
import sys, os, warnings
import numpy as np
import pandas as pd
from sklearn.model_selection import GroupKFold, GroupShuffleSplit, cross_validate
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, IsolationForest
from sklearn.dummy import DummyClassifier
from sklearn.neighbors import LocalOutlierFactor
from sklearn.inspection import permutation_importance

warnings.filterwarnings("ignore")
RS = 42

# ----------------------------- CONFIG --------------------------------------
FEATURES = ["pH", "DO", "BOD", "COD", "Cond", "Nitrate", "TotalColi",
            "FecalColi", "Turbidity", "TotAlk", "Chloride", "Sulphate",
            "Hardness", "TDS", "TSS", "Fluoride"]

TARGETS = ["wqi_class_4", "cpcb_class", "potable", "potability_class"]

# Columns that DEFINE each label (removed from that model's inputs to stop the
# model just relearning the rule). Adjust to match how your labels were built.
LABEL_DEFINING = {
    "wqi_class_4":      ["pH", "DO", "BOD", "Cond", "Nitrate"],
    "cpcb_class":       ["pH", "DO", "BOD", "FecalColi"],
    "potable":          ["pH", "TDS", "Hardness", "Chloride", "Sulphate", "Fluoride", "Nitrate"],
    "potability_class": ["pH", "TDS", "Hardness", "Chloride", "Sulphate", "Fluoride", "Nitrate"],
}

# first name found in the file is used
STATION_CANDIDATES = ["station", "Station", "STATION", "station_id", "station_code", "Station_Code", "STN_CODE"]
FLAG_CANDIDATES = ["review_group"]          # extreme_valid / suspect_error tags
LOG_COLS = ["BOD", "COD", "Cond", "Nitrate", "TotalColi", "FecalColi", "Turbidity",
            "TSS", "Chloride", "Sulphate", "Hardness", "TDS", "Fluoride", "TotAlk"]
CONTAMINATION = 0.05
# ---------------------------------------------------------------------------


def first_present(df, names):
    return next((c for c in names if c in df.columns), None)


def log_transform(X):
    return np.log1p(np.clip(X, 0, None))


def build_pipe(model, cols):
    log_idx = [i for i, c in enumerate(cols) if c in LOG_COLS]
    pre = []
    if log_idx:
        from sklearn.compose import ColumnTransformer
        ct = ColumnTransformer([("log", FunctionTransformer(log_transform), log_idx)],
                               remainder="passthrough")
        pre.append(("log", ct))
    return Pipeline([("imp", SimpleImputer(strategy="median"))] + pre +
                    [("sc", StandardScaler()), ("m", model)])


def supervised(df, groups, out):
    rows = []
    for t in TARGETS:
        if t not in df.columns:
            print(f"[skip] target {t} not in file")
            continue
        cols = [c for c in FEATURES if c in df.columns and c not in LABEL_DEFINING.get(t, [])]
        X, y = df[cols], df[t]
        cv = GroupKFold(n_splits=5)
        models = {
            "majority_baseline": DummyClassifier(strategy="most_frequent"),
            "logreg": LogisticRegression(max_iter=2000, class_weight="balanced"),
            "random_forest": RandomForestClassifier(
                n_estimators=300, max_depth=8, min_samples_leaf=5,
                class_weight="balanced", random_state=RS, n_jobs=-1),
        }
        for name, m in models.items():
            r = cross_validate(build_pipe(m, cols), X, y, groups=groups, cv=cv,
                               scoring=["accuracy", "f1_macro"], return_train_score=True)
            rows.append(dict(target=t, model=name, n_inputs=len(cols),
                             train_acc=r["train_accuracy"].mean(),
                             cv_acc=r["test_accuracy"].mean(),
                             cv_f1_macro=r["test_f1_macro"].mean(),
                             gap=r["train_accuracy"].mean() - r["test_accuracy"].mean()))
    res = pd.DataFrame(rows).round(3)
    print("\n=== Supervised results (station-grouped 5-fold CV) ===")
    print(res.to_string(index=False))
    print("A large 'gap' (train_acc - cv_acc) means overfitting; cv_acc near the"
          " majority baseline means underfitting / no signal.")
    res.to_csv(os.path.join(out, "supervised_results.csv"), index=False)


def importance(df, groups, out, target="potable"):
    if target not in df.columns:
        return
    cols = [c for c in FEATURES if c in df.columns and c not in LABEL_DEFINING.get(target, [])]
    tr, te = next(GroupShuffleSplit(test_size=0.25, random_state=RS).split(df, groups=groups))
    pipe = build_pipe(RandomForestClassifier(n_estimators=300, max_depth=8,
                      min_samples_leaf=5, class_weight="balanced", random_state=RS, n_jobs=-1), cols)
    pipe.fit(df.iloc[tr][cols], df.iloc[tr][target])
    pi = permutation_importance(pipe, df.iloc[te][cols], df.iloc[te][target],
                                n_repeats=20, random_state=RS, scoring="f1_macro")
    imp = pd.DataFrame({"feature": cols, "importance": pi.importances_mean,
                        "std": pi.importances_std}).sort_values("importance", ascending=False).round(4)
    print(f"\n=== Permutation importance for '{target}' (held-out stations) ===")
    print(imp.to_string(index=False))
    imp.to_csv(os.path.join(out, f"importance_{target}.csv"), index=False)


def anomalies(df, flag_col, out):
    cols = [c for c in FEATURES if c in df.columns]
    X = SimpleImputer(strategy="median").fit_transform(df[cols])
    li = [i for i, c in enumerate(cols) if c in LOG_COLS]
    X[:, li] = np.log1p(np.clip(X[:, li], 0, None))
    X = StandardScaler().fit_transform(X)

    iso = IsolationForest(n_estimators=400, contamination=CONTAMINATION, random_state=RS).fit(X)
    df = df.copy()
    df["iso_flag"] = (iso.predict(X) == -1).astype(int)
    df["iso_score"] = -iso.score_samples(X)                  # higher = more unusual
    lof = LocalOutlierFactor(n_neighbors=25, contamination=CONTAMINATION)
    df["lof_flag"] = (lof.fit_predict(X) == -1).astype(int)
    df["both_flag"] = df["iso_flag"] & df["lof_flag"]
    print(f"\n=== Anomaly detection ===\nIsolationForest flagged {df.iso_flag.sum()}, "
          f"LOF {df.lof_flag.sum()}, both {df.both_flag.sum()} of {len(df)}")

    if flag_col:   # false-alarm investigation: flagged vs 'valid extreme' tags
        print(f"\nIsolationForest flag vs '{flag_col}':")
        print(pd.crosstab(df[flag_col], df["iso_flag"]))
    if "potable" in df.columns:
        print("\nIsolationForest flag vs potable (1 = drinkable):")
        print(pd.crosstab(df["potable"], df["iso_flag"], normalize="index").round(3))
    df.to_csv(os.path.join(out, "anomaly_results.csv"), index=False)
    return df


def compare_with_supervised(df, groups, out, target="potable"):
    """Are anomalies just the samples the classifier gets wrong / the 'bad' water?"""
    from sklearn.model_selection import cross_val_predict
    if target not in df.columns or "iso_flag" not in df.columns:
        return
    cols = [c for c in FEATURES if c in df.columns and c not in LABEL_DEFINING.get(target, [])]
    pipe = build_pipe(RandomForestClassifier(n_estimators=300, max_depth=8, min_samples_leaf=5,
                      class_weight="balanced", random_state=RS, n_jobs=-1), cols)
    pred = cross_val_predict(pipe, df[cols], df[target], groups=groups, cv=GroupKFold(5))
    df = df.assign(pred=pred, wrong=(pred != df[target]).astype(int))
    print(f"\n=== Anomaly flag vs supervised errors ({target}) ===")
    print(pd.crosstab(df["iso_flag"], df["wrong"], rownames=["anomaly"], colnames=["misclassified"]))
    print("Misclassification rate among anomalies vs normal:",
          df.groupby("iso_flag")["wrong"].mean().round(3).to_dict())
    df.to_csv(os.path.join(out, "anomaly_vs_supervised.csv"), index=False)


def main(path, out="outputs"):
    os.makedirs(out, exist_ok=True)
    df = pd.read_csv(path)
    print("Rows:", len(df), "| columns:", len(df.columns))
    st = first_present(df, STATION_CANDIDATES)
    if st is None:
        raise SystemExit(f"No station column found. Add yours to STATION_CANDIDATES. Columns: {list(df.columns)}")
    groups = df[st].astype(str)
    print("Station column:", st, "| stations:", groups.nunique())
    missing = [c for c in FEATURES if c not in df.columns]
    if missing:
        print("[warn] feature columns not found:", missing)

    supervised(df, groups, out)
    importance(df, groups, out, "potable")
    flagged = anomalies(df, first_present(df, FLAG_CANDIDATES), out)
    compare_with_supervised(flagged, groups, out, "potable")
    print(f"\nSaved results to {out}/")


if __name__ == "__main__":
    # In Jupyter, sys.argv holds the kernel file, so call main() yourself:
    #     main("nwmp_model_ready.csv", "outputs")
    in_notebook = "ipykernel" in sys.modules
    if in_notebook:
        print("Notebook detected. Run:  main(r'path/to/nwmp_model_ready.csv', 'outputs')")
    else:
        if len(sys.argv) < 2:
            raise SystemExit(__doc__)
        main(sys.argv[1], sys.argv[2] if len(sys.argv) > 2 else "outputs")
```

```python
"""
Step 2: train the final models on ALL rows and save them for the web app.

Usage:  python train_export.py nwmp_model_ready.csv
In Jupyter:  train_export(r"path/to/nwmp_model_ready.csv")

Creates a  model/  folder (models + meta.json) that app.py loads.
Needs: pandas numpy scikit-learn joblib
"""
import sys, os, json, warnings
import numpy as np
import pandas as pd
import joblib
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, FunctionTransformer
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestClassifier, IsolationForest

from wq_utils import log_transform  # saved models refer to this module by name

warnings.filterwarnings("ignore")
RS = 42

# ----------------------------- CONFIG (same as nwmp_models.py) -------------
FEATURES = ["pH", "DO", "BOD", "COD", "Cond", "Nitrate", "TotalColi",
            "FecalColi", "Turbidity", "TotAlk", "Chloride", "Sulphate",
            "Hardness", "TDS", "TSS", "Fluoride"]
UNITS = {"pH": "", "DO": "mg/L", "BOD": "mg/L", "COD": "mg/L", "Cond": "µS/cm",
         "Nitrate": "mg/L as N", "TotalColi": "MPN/100mL", "FecalColi": "MPN/100mL",
         "Turbidity": "NTU", "TotAlk": "mg/L", "Chloride": "mg/L", "Sulphate": "mg/L",
         "Hardness": "mg/L", "TDS": "mg/L", "TSS": "mg/L", "Fluoride": "mg/L"}
TARGETS = ["wqi_class_4", "cpcb_class", "potable"]
LABEL_DEFINING = {
    "wqi_class_4": ["pH", "DO", "BOD", "Cond", "Nitrate"],
    "cpcb_class":  ["pH", "DO", "BOD", "FecalColi"],
    "potable":     ["pH", "TDS", "Hardness", "Chloride", "Sulphate", "Fluoride", "Nitrate"],
}
LOG_COLS = ["BOD", "COD", "Cond", "Nitrate", "TotalColi", "FecalColi", "Turbidity",
            "TSS", "Chloride", "Sulphate", "Hardness", "TDS", "Fluoride", "TotAlk"]
CONTAMINATION = 0.05
# hard physical limits: a value outside these is almost surely a data-entry error
PHYSICAL = {"pH": (0, 14), "DO": (0, 20), "BOD": (0, 5000), "Cond": (0, 100000),
            "Turbidity": (0, 5000), "TDS": (0, 60000)}
# ---------------------------------------------------------------------------


def make_pipe(model, cols):
    log_idx = [i for i, c in enumerate(cols) if c in LOG_COLS]
    steps = [("imp", SimpleImputer(strategy="median"))]
    if log_idx:
        steps.append(("log", ColumnTransformer(
            [("l", FunctionTransformer(log_transform), log_idx)], remainder="passthrough")))
    steps += [("sc", StandardScaler()), ("m", model)]
    return Pipeline(steps)


def train_export(path, out="model"):
    os.makedirs(out, exist_ok=True)
    df = pd.read_csv(path)
    feats = [c for c in FEATURES if c in df.columns]
    print("Rows:", len(df), "| features used:", len(feats))

    meta = {"features": feats, "units": {c: UNITS.get(c, "") for c in feats},
            "targets": {}, "physical": PHYSICAL, "stats": {}}
    for c in feats:
        s = df[c].dropna()
        meta["stats"][c] = {"min": float(s.min()), "max": float(s.max()),
                            "p01": float(s.quantile(.01)), "p99": float(s.quantile(.99)),
                            "median": float(s.median())}

    # supervised models (each hides its own label-defining columns)
    for t in TARGETS:
        if t not in df.columns:
            print("[skip]", t); continue
        cols = [c for c in feats if c not in LABEL_DEFINING[t]]
        pipe = make_pipe(RandomForestClassifier(
            n_estimators=300, max_depth=8, min_samples_leaf=5,
            class_weight="balanced", random_state=RS, n_jobs=-1), cols)
        pipe.fit(df[cols], df[t])
        joblib.dump(pipe, os.path.join(out, f"{t}.joblib"))
        meta["targets"][t] = {"inputs": cols, "classes": [str(c) for c in pipe.classes_]}
        print("saved", t, "| inputs:", len(cols))

    # anomaly model uses all features
    iso = make_pipe(IsolationForest(n_estimators=400, contamination=CONTAMINATION,
                                    random_state=RS), feats)
    iso.fit(df[feats])
    train_scores = -iso.score_samples(df[feats])
    meta["anomaly"] = {"threshold": float(np.quantile(train_scores, 1 - CONTAMINATION)),
                       "train_scores": np.round(np.sort(train_scores)[::10], 4).tolist()}
    joblib.dump(iso, os.path.join(out, "anomaly.joblib"))
    with open(os.path.join(out, "meta.json"), "w") as f:
        json.dump(meta, f, indent=1)
    print("Done. Now run:  python app.py")


if __name__ == "__main__":
    if "ipykernel" in sys.modules:
        print("Notebook detected. Run:  train_export(r'path/to/nwmp_model_ready.csv')")
    elif len(sys.argv) < 2:
        raise SystemExit(__doc__)
    else:
        train_export(sys.argv[1])
```
